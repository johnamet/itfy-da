# Topic 19d Appendix - Submitting your work with GitHub

## A step-by-step guide for Windows

You do not need any previous experience with Git or GitHub to follow this guide. Everything is explained from the beginning, and the main route uses **GitHub Desktop**, an application with buttons instead of typed commands.

Read this guide once from top to bottom before you start. Then keep it open beside you and follow it step by step while you submit.

Every assignment in this course is submitted the same way, so you only have to learn this once. Appendix A at the end shows the same steps as typed commands, because you will meet those commands in real workplaces.

## What you will achieve

By the end of this guide you will have:

1. a free GitHub account;
2. GitHub Desktop installed and signed in on your Windows computer;
3. a public repository containing your completed notebook, your data files, and a README;
4. a link you can paste into your submission form;
5. the ability to correct and re-submit your work if you find a mistake.

Time needed: about 30 minutes the first time, and about 5 minutes for every assignment after that.

## 1. Why analysts use GitHub

Imagine you finish an analysis, email `analysis_final.ipynb` to a colleague, then fix a mistake and email `analysis_final_v2.ipynb`. A week later somebody asks which file produced the chart in the report. Nobody knows.

**Version control** solves this. It is a system that records every saved change to a project, who made it, when, and why. **Git** is the version control program. **GitHub** is a website that stores Git projects online so other people can see them.

For a data analyst this matters for three reasons:

| Reason | What it means for you |
|---|---|
| Reproducibility | Your notebook, your data, and your notes stay together, so somebody else can re-run your analysis and get the same answer |
| History | If you break something, the earlier working version is still recorded |
| Evidence | A public link proves what you produced and when, which is exactly what an employer wants to see |

This last point is worth taking seriously. The repository you create for this assignment is a piece of portfolio work that belongs to you. Write it as though a future employer will read it, because one might.

## 2. The words you need

Learn these seven words now. The rest of the guide uses them constantly.

| Word | Plain meaning |
|---|---|
| **Repository** (or *repo*) | A project folder that Git is watching. It holds your files and their whole history. |
| **Commit** | A saved snapshot of your project at one moment, with a short message explaining what changed. Think of it as a labelled save point. |
| **Commit message** | The short note you write with each commit, for example `Add cleaned data and line chart`. |
| **Remote** | The copy of your repository that lives on GitHub's servers, as opposed to the copy on your computer. |
| **Push** | Send your commits from your computer up to GitHub. |
| **Pull** | Bring changes from GitHub down to your computer. |
| **README.md** | A markdown file that GitHub displays on your repository's front page. It is the first thing a reader sees. |

Two more you will see mentioned but do not need for this assignment: a **branch** is a separate line of work, and a **clone** is a downloaded copy of an existing repository.

> **A note on saving.** Saving your notebook in VS Code or Jupyter does *not* create a commit, and creating a commit does *not* put your work on GitHub. There are three separate actions: **save** the file, **commit** the change, **push** to GitHub. Your work is only submitted after the third one.

## 3. One-time setup on Windows

You only do this section once, ever. If you have already done it for a previous assignment, skip to section 4.

### Step 3.1 - Create a GitHub account

1. Open a browser and go to **https://github.com/signup**.
2. Enter an email address you can access. Use a professional-looking one if you have it.
3. Choose a password and write it down somewhere safe.
4. Choose a username. **Pick this carefully** - it appears in every link you ever share, so `ama-mensah` is a better choice than `xX_coder_Xx`. Use letters, numbers, and hyphens only.
5. Verify your email address by clicking the link GitHub sends you.
6. When GitHub asks about a plan, choose the **free** option. You never need to pay for this course.

### Step 3.2 - Install GitHub Desktop

1. Go to **https://desktop.github.com**.
2. Click **Download for Windows**. The file is named something like `GitHubDesktopSetup-x64.exe`.
3. Open the downloaded file. It installs itself and opens automatically - there are no options to choose.
4. If Windows shows a security warning saying the publisher could not be verified, check that you downloaded from `desktop.github.com` and then allow it.

GitHub Desktop includes Git, so you do **not** need to install Git separately.

### Step 3.3 - Sign in and set your name

1. In GitHub Desktop, click **Sign in to GitHub.com** and enter the account details from step 3.1. Your browser will open to confirm, then return you to the app.
2. GitHub Desktop then shows a **Configure Git** screen with a name and an email address. These are stamped onto every commit you make so that your work is attributed to you. Leave the values it suggests unless they are wrong, then click **Finish**.

You are now set up. Everything from here is what you repeat for each assignment.

## 4. Create the repository for this assignment

### Step 4.1 - Create it

1. In GitHub Desktop, open the **File** menu and choose **New repository...**.
2. Fill in the form:

| Field | What to put |
|---|---|
| **Name** | `itfy-data-gathering-download` for the download assignment, or `itfy-data-gathering-api` for the API assignment. Add your own name if you prefer, for example `itfy-data-gathering-api-ama`. Use hyphens, not spaces. |
| **Description** | One sentence, for example `Data gathering assignment for the ITFY Ghana data analytics course.` |
| **Local path** | Leave the suggested path. It is usually `C:\Users\<YourName>\Documents\GitHub`. Note it down - you need it in step 5. |
| **Initialize this repository with a README** | **Tick this box.** |
| **Git ignore** | Choose **Python** from the dropdown. |
| **License** | Leave as **None**. |

3. Click **Create repository**.

Choosing the **Python** git ignore matters. It tells Git to ignore files that should never be uploaded, such as your `.venv` virtual environment folder and `__pycache__` folders. Without it you would accidentally try to upload thousands of files.

### Step 4.2 - Publish it to GitHub

Your repository currently exists only on your computer. To put it online:

1. Click the **Publish repository** button at the top of GitHub Desktop.
2. Keep the name as it is.
3. **Untick "Keep this code private".** Your repository must be public so that your tutor can open it without being invited.
4. Click **Publish repository**.

Your repository now exists at `https://github.com/<your-username>/<repository-name>`.

## 5. Put your work into the repository folder

This is the step most people find confusing, so read it slowly. GitHub Desktop is only watching **one specific folder on your computer**. If your notebook is not inside that folder, GitHub Desktop cannot see it.

### Step 5.1 - Open the folder

In GitHub Desktop, open the **Repository** menu and choose **Show in Explorer**. A File Explorer window opens showing your repository folder. It is usually at:

```text
C:\Users\<YourName>\Documents\GitHub\itfy-data-gathering-api
```

Right now it contains a `README.md` and a `.gitignore` file. The `.git` folder is hidden and you should never touch it.

### Step 5.2 - Build this structure

Copy your completed notebook into that folder, and create the subfolders shown below. In File Explorer you create a folder with **right-click, New, Folder**.

```text
itfy-data-gathering-api\
    README.md                              <- you will write this in section 7
    .gitignore                             <- created for you, leave it alone
    topic19_assignment_api_gathering.ipynb <- your completed notebook
    data\
        raw\
            open_meteo_accra_raw.json      <- the untouched response you saved
        clean\
            accra_daily_weather_clean.csv  <- the cleaned table you saved
    charts\
        daily_temperature_line.png         <- optional, only if you saved images
```

The exact filenames depend on which dataset you chose. The structure is what matters: **the notebook at the top, raw data separated from clean data.** That separation is the same principle taught in the data gathering lesson - the raw file is evidence of what you received, and the clean file is your interpretation of it.

### Step 5.3 - Check your file sizes

GitHub refuses any single file over 100 MB and warns above 50 MB. The datasets in these assignments are all under 1 MB, so you should have no trouble. To check, right-click a file, choose **Properties**, and read the Size line.

If a file is unexpectedly large, do not upload it. Instead, add its name to `.gitignore` and explain in your README where the file came from and how to download it again. Appendix B shows how.

## 6. Commit and push your work

Now you save a snapshot and send it to GitHub.

### Step 6.1 - Look at what changed

Switch back to GitHub Desktop and click the **Changes** tab on the left. You should see a list of your new files, each with a tick beside it. Click any file to see what Git noticed about it.

If you see nothing, your files are not in the repository folder. Go back to step 5.1.

> Notebooks look strange in this view. GitHub Desktop shows the raw `.ipynb` file, which is JSON text, so you will see lines of `"cell_type": "markdown"` rather than your neat notebook. This is normal and it is not a problem.

### Step 6.2 - Write a commit message

At the bottom left there are two boxes. In the **Summary** box write a short description of what you are saving. Good messages say what changed:

| Good | Poor |
|---|---|
| `Add completed API gathering assignment` | `stuff` |
| `Add cleaned data and monthly rainfall bar chart` | `update` |
| `Fix year column type in cleaning function` | `asdf` |

Keep it under about 60 characters and describe the change, not your feelings about it. Leave the larger Description box empty unless you have more to explain.

### Step 6.3 - Commit, then push

1. Click the blue **Commit to main** button. Your changes are now recorded on your computer.
2. Click **Push origin** at the top of the window. Your commit is now on GitHub.

Both steps are required. A commit without a push means your work is saved locally but nobody else can see it - which means it is not submitted.

## 7. Write your README

Your README is the front page of your repository and part of what you are marked on. GitHub displays it automatically, so it is where your reader learns what you did.

Edit `README.md` in the repository folder. You can open it in VS Code, or in Notepad, or edit it directly on github.com by clicking the file and then the pencil icon.

Copy this template and replace everything in square brackets:

```markdown
# [Assignment title, for example: Data Gathering - API Assignment]

**Name:** [Your full name]
**Course:** ITFY Ghana - Data Analytics with Python
**Submitted:** [Date]

## Research question

[One or two sentences stating exactly what you set out to find out.]

## Data source

| Item | Detail |
|---|---|
| Publisher | [Who publishes this data] |
| URL or endpoint | [The exact address you requested] |
| Retrieved on | [Date you downloaded it] |
| What one row represents | [For example: one country in one year] |
| Licence or terms | [Link to the terms of use] |

## How to run this notebook

1. Install the requirements: `pip install pandas matplotlib requests`
2. Open `[your-notebook-name].ipynb`
3. Run all cells from top to bottom.

The raw response is saved in `data/raw/` and the cleaned table in `data/clean/`,
so the analysis can be re-run without downloading the data again.

## What I found

- [Finding one, with a number in it]
- [Finding two, with a number in it]
- [Finding three, with a number in it]

## Limitations

[Two or three sentences on what this data cannot tell you, and why.]
```

Write findings that contain numbers. "Life expectancy increased" is weak; "life expectancy in Ghana rose from 45.7 years in 1960 to 63.8 in 2021" is evidence.

After editing, commit and push again using section 6. Every change follows that same loop: **edit, commit, push.**

## 8. Check your work, then submit the link

### Step 8.1 - Look at it the way your tutor will

1. In GitHub Desktop, open the **Repository** menu and choose **View on GitHub**. Your browser opens your repository.
2. Check that your README appears on the front page and reads correctly.
3. Click your `.ipynb` file. GitHub renders notebooks, so you should see your markdown, your code, **and your chart images**.
4. Click into `data/raw` and `data/clean` and confirm both files are there.

> **If your charts do not appear**, you saved the notebook without its outputs. Open the notebook again, run every cell from top to bottom, save it, then commit and push. GitHub can only display outputs that were saved inside the file.

### Step 8.2 - Prove it is public

This catches the single most common submission failure. Copy your repository URL, then open a **private** or **incognito** browser window (in Edge or Chrome, press `Ctrl` + `Shift` + `N`) and paste it there.

- If the page loads, your repository is public and your tutor can read it.
- If you see a 404 page, it is still private. Go to your repository on GitHub, click **Settings**, scroll to the bottom, and under **Danger Zone** choose **Change repository visibility**, then make it public.

### Step 8.3 - Submit

Submit this link, replacing the placeholders with your own username and repository name:

```text
https://github.com/<your-username>/<repository-name>
```

Do not submit a link that contains `/blob/main/` or ends in `.ipynb` - that points at a single file. Do not submit a screenshot. The plain repository link is what is wanted.

## 9. Fixing a mistake after you have submitted

You will find a mistake after submitting. Everybody does, and it is not a problem, because your link always shows the latest version.

1. Fix the notebook on your computer and save it.
2. Open GitHub Desktop. Your change appears in the **Changes** tab.
3. Write a commit message describing the fix, for example `Correct axis label on rainfall chart`.
4. Click **Commit to main**, then **Push origin**.

Refresh your repository page and the correction is there. Your commit history now shows both the original work and the fix, which is exactly how professional work looks. Do not delete the repository and start again - the history is a feature, not a mess.

## Submission checklist

Work through this before you submit. Every line is something tutors regularly find missing.

- [ ] The repository is **public** and opens in a private browsing window
- [ ] The repository name follows the pattern given in the assignment
- [ ] The completed `.ipynb` notebook is in the repository
- [ ] The notebook was run from top to bottom before saving, so outputs and charts are visible on GitHub
- [ ] `data/raw/` contains the untouched downloaded file or saved API response
- [ ] `data/clean/` contains the cleaned dataset you saved
- [ ] `README.md` is complete, with the research question, source table, findings, and limitations
- [ ] Every commit has a message that describes the change
- [ ] There is **no** API key, password, token, or personal data anywhere in the repository
- [ ] `.venv` and `__pycache__` are **not** in the repository
- [ ] The final commit has been **pushed**, and GitHub Desktop shows no unpushed changes

## Appendix A - the same steps as typed commands

GitHub Desktop is doing ordinary Git commands for you behind its buttons. Learning the commands is worthwhile because servers and workplaces often have no graphical interface. This appendix is optional for the assignment and required knowledge for a job.

On Windows, GitHub Desktop installs a terminal called **Git Bash**. Find it from the Start menu, or right-click inside your repository folder and choose **Open Git Bash here** if the option appears. In VS Code you can use the built-in terminal instead.

### Button to command

| GitHub Desktop button | Git command |
|---|---|
| Configure Git (name and email) | `git config --global user.name "Your Name"` and `git config --global user.email "you@example.com"` |
| File, New repository | `git init` |
| Show in Explorer | `explorer .` |
| The Changes tab | `git status` |
| The ticks beside each file | `git add .` |
| Commit to main | `git commit -m "Add completed assignment"` |
| Publish repository | `git remote add origin <url>` then `git push -u origin main` |
| Push origin | `git push` |
| Fetch or Pull origin | `git pull` |
| History tab | `git log --oneline` |

### A full submission in commands

Run these inside your repository folder, one line at a time:

```bash
git status
git add .
git commit -m "Add completed API gathering assignment"
git push
```

`git status` first, every time. It tells you which files Git has noticed before you commit anything, and it is the command that prevents most mistakes.

### Cloning an existing repository

To download a repository that already exists on GitHub:

```bash
git clone https://github.com/<username>/<repository-name>.git
cd <repository-name>
```

### A warning about `git add .`

The dot means "everything in this folder". It is convenient and it is also how people accidentally commit passwords and 2 GB data files. Run `git status` first and read the list. To add one file only, name it: `git add topic19_assignment_api_gathering.ipynb`.

## Appendix B - Windows troubleshooting

### "I cannot see my notebook in GitHub Desktop"

Your file is not inside the repository folder. In GitHub Desktop use **Repository, Show in Explorer** to open the correct folder, then move your notebook into it. Having a copy elsewhere on your computer changes nothing.

### "My charts are missing on GitHub"

The notebook was saved without outputs. Run all cells, save, then commit and push. In VS Code, use **Run All**, wait for every cell to finish, then press `Ctrl` + `S`.

### "This file is too large" or "file exceeds GitHub's file size limit"

GitHub rejects files over 100 MB. Open `.gitignore` in Notepad or VS Code and add a line naming the file or folder, for example:

```text
data/raw/very_large_file.csv
```

Then explain in your README where that file came from and how to download it again. Provenance in words is an acceptable substitute for a file that will not fit; a missing explanation is not.

### "Authentication failed" or it keeps asking for a password

Sign out and back in: in GitHub Desktop open **File, Options, Accounts**, sign out, then sign in again. GitHub stopped accepting account passwords for Git operations in 2021, so if a command line asks you for a password, it wants a personal access token rather than the password you log in with. Using GitHub Desktop avoids this entirely.

### "warning: LF will be replaced by CRLF"

Not an error. Windows and Linux mark the end of a line differently, and Git is telling you it is converting between them. Ignore it.

### My repository folder is inside OneDrive

OneDrive and Git both try to manage the same files and sometimes fight, which shows up as files that lock or revert. If your `Documents` folder is synced to OneDrive, create your repositories somewhere outside it instead, for example `C:\dev\`. In GitHub Desktop you can set the **Local path** when you create the repository.

### Paths with spaces

A folder named `My Data Files` will break command line paths unless you quote it: `cd "C:\Users\Ama\My Data Files"`. Save yourself the trouble and avoid spaces in folder names for anything programming-related.

### "I committed a password or an API key"

Tell your tutor, and treat the secret as compromised: go to the service that issued it and revoke or regenerate it immediately. Deleting the line and committing again is **not** enough, because the old commit still contains it and the history is public. This is why the assignments tell you to keep keys in environment variables rather than typing them into a notebook.

### I deleted something by accident

If you have not committed yet, GitHub Desktop can undo it: right-click the change in the **Changes** tab and choose **Discard changes**. If you have already committed, the earlier version is in the **History** tab - open it and ask your tutor for help recovering the file rather than experimenting.
