## What to Look For

Good dataset documentation usually answers:

| Documentation item | Why it matters |
|--------------------|----------------|
| Dataset title | Helps identify the exact dataset |
| Publisher/source | Helps judge trust and citation |
| Date downloaded / updated | Tells whether the data is current |
| Row meaning | Prevents wrong interpretation |
| Column definitions | Explains variables clearly |
| Units | Prevents wrong calculations |
| Time coverage | Shows what period can be analysed |
| Geographic coverage | Shows where the data applies |
| Missing-value rules | Explains blanks, nulls, or special codes |
| Processing notes | Reveals transformations already applied |
| License | Tells whether and how you may reuse the data |


```python
import json
import pandas as pd

METADATA_PATH = "investigate_dataset/data/owid_life_expectancy_metadata.json"

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

print(metadata.keys())
```

```python
chart = metadata["chart"]
variable_meta = list(metadata["columns"].values())[0]

metadata_summary = {
    "title": chart.get("title"),
    "citation": chart.get("citation"),
    "original_url": chart.get("originalChartUrl"),
    "unit": variable_meta.get("unit"),
    "timespan": variable_meta.get("timespan"),
    "type": variable_meta.get("type"),
    "last_updated": variable_meta.get("lastUpdated"),
    "next_update": variable_meta.get("nextUpdate"),
}

for key, value in metadata_summary.items():
    print(f"{key:15s}: {value}")
```

---
## Turn Metadata into Analysis Notes

Before analysis, write a short note like this:

> This dataset measures life expectancy in years for countries and regions over time. It is published through Our World in Data and combines multiple source datasets. Because the data has been processed from several sources, I should cite OWID and avoid claiming that all values were collected by one organisation using one method.

This kind of note protects your analysis from overclaiming.
