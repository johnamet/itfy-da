This appendix lists places students can look for datasets and what to check before using them.

## Dataset Source Guide

| Source | Good for | Strengths | Watch out for |
|--------|----------|-----------|---------------|
| Kaggle | Practice datasets, competitions, examples | Easy search, notebooks, many CSVs | Quality varies; some datasets lack documentation |
| Our World in Data | Health, population, climate, development | Clear metadata, citations, CSV downloads | Variables may combine multiple sources |
| World Bank DataBank | Country indicators, development economics | Official global data, API access | Can be overwhelming for beginners |
| data.gov | Public US government data | Government source, many topics | Formats and quality vary by agency |
| Ghana Statistical Service | Ghana census and official statistics | Best official Ghana source | Some files may be PDFs or reports, not tidy CSVs |
| Google Dataset Search | Finding datasets across websites | Broad discovery tool | Not a publisher; still verify the original source |
| GitHub | Project datasets and open data repos | Easy raw file downloads | Check README, license, and maintenance |

A good dataset for a beginner project should have clear columns, enough rows to analyse, documentation, and permission to use it.


---
## Quick Source Evaluation Questions

Before choosing a dataset, ask:

1. Who published it?
2. Why was it collected?
3. When was it last updated?
4. What does each row represent?
5. Are column definitions provided?
6. Is there a license or terms of use?
7. Is the file format usable for this course?
8. Are there enough rows for analysis?
9. Are there obvious quality issues?
10. Can I cite the source clearly?
