Before analysis begins, a good analyst asks:

- Where did this data come from?
- Who collected or published it?
- What does each column mean?
- What is the unit of measurement?
- What time period does it cover?
- Are there limitations, missing values, or processing decisions?

This lesson teaches the pre-analysis stage: finding a dataset, reading its documentation, and understanding it before writing analysis code.

**Related analysis lesson:** [Topic 19 — Global Development Data Investigation](topic19_global_development_analysis.ipynb)


---
## 🎯 Why This Matters

Beginners often open a CSV and immediately start calculating. That is risky.

If you do not understand the dataset first, you may:

- use the wrong unit
- compare values that should not be compared
- miss important limitations
- misunderstand how missing values were handled
- make claims the data cannot support

Professional analysis starts with **source understanding** and **data documentation**, not charts.


---
## 📚 Common Places to Gather Data

| Source | Best for | Notes |
|--------|----------|-------|
| Kaggle | Practice datasets, competitions, notebooks | Often easy to browse, but check licenses and dataset quality |
| Our World in Data | Global development, health, climate, population | Usually includes metadata, citations, and downloadable CSVs |
| World Bank DataBank / API | Country indicators and development data | Official source for many economic and social indicators |
| data.gov | United States public datasets | Good for government datasets, varies by agency |
| Ghana Statistical Service | Ghana census, surveys, official statistics | Best source for Ghana-specific official statistics |
| Google Dataset Search | Finding datasets across the web | Search tool, not a data publisher itself |
| GitHub repositories | Public project datasets | Quality varies; check README, license, and source |

The source matters because it affects trust, documentation, update frequency, and whether you are allowed to reuse the data.

## 🧪 Worked Example: Our World in Data

Our World in Data makes many chart datasets downloadable as CSV files.

For many OWID charts, you can add `.csv` or `.metadata.json` to the chart URL:

```text
https://ourworldindata.org/grapher/life-expectancy.csv
https://ourworldindata.org/grapher/life-expectancy.metadata.json
```

In this lesson folder, both files have already been downloaded:

- `investigate_dataset/data/owid_life_expectancy.csv`
- `investigate_dataset/data/owid_life_expectancy_metadata.json`


````python
import json
import pandas as pd

DATA_PATH = "investigate_dataset/data/owid_life_expectancy.csv"
METADATA_PATH = "investigate_dataset/data/owid_life_expectancy_metadata.json"

life = pd.read_csv(DATA_PATH)

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

print("Data loaded:", life.shape)
print(life.head())
````

---
## 🔎 Step 1 — Read the Dataset Shape and Columns

Before analysis, ask: what does one row represent?


````python
print("Shape:", life.shape)
print("Columns:", list(life.columns))
print()
print("Data types:")
print(life.dtypes)
print()
print("First rows:")
print(life.head())
````

In this dataset, one row represents one place (`Entity`) in one year (`Year`) with a life expectancy value.


---
## 📖 Step 2 — Read the Metadata

Metadata is data about the data. It tells us what the dataset means and how to cite it.


````python
chart = metadata["chart"]
columns = metadata["columns"]

print("Chart title:", chart.get("title"))
print("Original chart URL:", chart.get("originalChartUrl"))
print("Citation:", chart.get("citation"))
print("Downloaded on:", metadata.get("dateDownloaded"))
print()
print("Metadata column keys:")
print(list(columns.keys()))
````

````python
variable_meta = list(metadata["columns"].values())[0]

print("Short title:", variable_meta.get("titleShort"))
print("Unit:", variable_meta.get("unit"))
print("Short unit:", variable_meta.get("shortUnit"))
print("Timespan:", variable_meta.get("timespan"))
print("Type:", variable_meta.get("type"))
print("Last updated:", variable_meta.get("lastUpdated"))
print("Next update:", variable_meta.get("nextUpdate"))
````

---
## 🧠 Step 3 — Translate Documentation into Plain English

A good analyst writes down their understanding before analysis.

Example:

| Question | Answer from metadata |
|----------|----------------------|
| What is measured? | Life expectancy at birth |
| Unit | Years |
| Time coverage | Long-run historical data through recent years |
| Source/citation | OWID compilation from Riley, Zijdeman, HMD, and UN WPP |
| Important caution | The chart combines multiple sources and includes processing decisions |

This prevents vague statements like "health got better" when the actual variable is life expectancy at birth.


````python
print("Description:")
print(variable_meta.get("descriptionShort"))
print()
print("Processing notes preview:")
processing = variable_meta.get("descriptionProcessing", "")
print(processing[:700])
````

## 🧹 Step 4 — First Data Understanding Checks

We are still not doing deep analysis. We are checking whether the dataset behaves as expected.


````python
print("Missing values:")
print(life.isna().sum())
print()
print("Duplicate rows:", life.duplicated().sum())
print("Entities:", life["Entity"].nunique())
print("Year range:", life["Year"].min(), "to", life["Year"].max())
````

````python
print("Example entities:")
print(life["Entity"].drop_duplicates().head(20).to_list())
print()
print("Rows for Ghana:")
print(life[life["Entity"] == "Ghana"].head())
````

---
## 🧾 Step 5 — Create a Mini Data Dictionary

A data dictionary is a table that explains each column.


````python
data_dictionary = pd.DataFrame([
    {"column": "Entity", "meaning": "Country, region, or aggregate entity", "example": life["Entity"].iloc[0]},
    {"column": "Code", "meaning": "Country/entity code where available", "example": life["Code"].dropna().iloc[0]},
    {"column": "Year", "meaning": "Year of observation", "example": life["Year"].iloc[0]},
    {"column": "Life expectancy", "meaning": "Expected years of life at birth", "example": life["Life expectancy"].iloc[0]},
])

print(data_dictionary)
````

## ✅ Pre-Analysis Checklist

Before starting any serious analysis, complete this checklist:

| Question | Checked? |
|----------|----------|
| Do I know the original source? | ☐ |
| Do I know what one row represents? | ☐ |
| Do I know what each column means? | ☐ |
| Do I know the units? | ☐ |
| Do I know the time range? | ☐ |
| Have I checked missing values and duplicates? | ☐ |
| Have I read limitations or processing notes? | ☐ |
| Do I know how to cite the dataset? | ☐ |

Once this checklist is complete, you are ready for analysis.
