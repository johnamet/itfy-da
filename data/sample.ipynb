```python
import matplotlib.pyplot as plt
import numpy as np

waffle_data = {
    "Football": 45,
    "Basketball": 25,
    "Volleyball": 20,
    "Athletics": 10
}

labels = list(waffle_data.keys())
values = list(waffle_data.values())

# Waffle chart dimensions (10 × 10 = 100 squares)
rows, cols = 10, 10

# Different hatch patterns for black-and-white printing
hatches = ['///', '\\\\\\', 'xxx', '...']

# Create figure
fig, ax = plt.subplots(figsize=(8, 8))

# Fill the waffle chart
square_num = 0

for i, (label, value) in enumerate(waffle_data.items()):
    for _ in range(value):
        row = rows - 1 - (square_num // cols)
        col = square_num % cols

        rect = plt.Rectangle(
            (col, row),
            1, 1,
            facecolor='white',
            edgecolor='black',
            hatch=hatches[i],
            linewidth=0.8
        )

        ax.add_patch(rect)
        square_num += 1

# Formatting
ax.set_xlim(0, cols)
ax.set_ylim(0, rows)
ax.set_aspect('equal')
ax.set_xticks([])
ax.set_yticks([])

# Legend
legend_handles = [
    plt.Rectangle((0, 0), 1, 1,
                  facecolor='white',
                  edgecolor='black',
                  hatch=hatches[i])
    for i in range(len(labels))
]

ax.legend(
    legend_handles,
    labels,
    loc='upper center',
    bbox_to_anchor=(0.5, -0.05),
    ncol=2
)

plt.title("Favourite Sports of Students", fontsize=14)
plt.tight_layout()

# Save for printing
plt.savefig(
    "waffle_chart_bw.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

weekly_savings = {
    10: 5,
    20: 10,
    30: 8,
    40: 12,
    50: 5
}

amount_saved = list(weekly_savings.keys())
frequency = list(weekly_savings.values())

# Create text to display inside sectors
def make_autopct(labels, values):
    total = sum(values)

    def autopct(pct):
        idx = autopct.counter
        angle = values[idx] * 360 / total
        text = f"GH₵{labels[idx]}\n{angle:.0f}°"
        autopct.counter += 1
        return text

    autopct.counter = 0
    return autopct

plt.figure(figsize=(8, 8))

plt.pie(
    frequency,
    colors=['white'] * len(frequency),  # no colours
    wedgeprops={
        'edgecolor': 'black',
        'linewidth': 1.5
    },
    startangle=90,
    labels=None,  # no labels outside
    autopct=make_autopct(amount_saved, frequency),
    pctdistance=0.65,  # move text inside sectors
    textprops={
        'fontsize': 10,
        'ha': 'center',
        'va': 'center'
    }
)

plt.title("Weekly Savings Distribution")
plt.axis('equal')

plt.savefig("pie_chart_bw.png", dpi=300, bbox_inches='tight')
plt.show()
```
