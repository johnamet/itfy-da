## ⚡ Part 1 — Why NumPy Exists

In Topic 14 you processed data with Python lists and for loops.
That works — but it has a serious speed problem.

Run the cell below and watch what happens when we try to process
a year of temperature readings both ways.
**Read more:** For the concept behind `import time` and later `import numpy as np`, see [Topic 15A — Importing Libraries](topic15a_importing_libraries.ipynb).


```python
import time

# Simulate a large dataset — 1 million temperature readings
big_list = [28.5 + (i % 10) * 0.3 for i in range(1_000_000)]

# ── METHOD 1: Python list + for loop ────────────────────────
start = time.time()
total = 0
for val in big_list:
    total += val
average_loop = total / len(big_list)
loop_time = time.time() - start

# ── METHOD 2: NumPy array ────────────────────────────────────
import numpy as np

big_array = np.array(big_list)

start = time.time()
average_numpy = np.mean(big_array)
numpy_time = time.time() - start

# ── Results ──────────────────────────────────────────────────
print(f"Python loop : {loop_time:.4f} seconds  →  avg = {average_loop:.2f}")
print(f"NumPy       : {numpy_time:.4f} seconds  →  avg = {average_numpy:.2f}")
print()
if loop_time > 0:
    speedup = round(loop_time / max(numpy_time, 0.0001), 0)
    print(f"NumPy is approximately {speedup}x faster on 1 million values.")
print()
print("Same result. NumPy wins on speed — every time.")
```

**Why is NumPy faster?**

A Python list can hold anything — strings, integers, booleans, mixed types.
That flexibility has a cost: Python must check the type of every item before
operating on it. A NumPy array holds **one type only** — it knows every element
is a float64 and applies the operation across all of them at the hardware level.

This is called **vectorisation** — operating on the whole array at once
instead of looping item by item. It is the engine beneath pandas, matplotlib,
and every serious data tool in Python.

## ⚡ Part 4 — Vectorised Operations

This is where NumPy's power really shows.
Any arithmetic operation on an array applies to **every element at once**.
No loop needed.
**Extra reading:** [Topic 15D — Vectorisation and Broadcasting](topic15d_vectorisation_broadcasting.ipynb) explains the deeper idea behind array-wide calculations.


```python
temps_celsius = np.array([28.5, 30.1, 27.8, 31.2, 29.4, 26.5, 32.0])

# ── Convert to Fahrenheit in one line ────────────────────────
# Formula: F = (C × 9/5) + 32
temps_fahrenheit = (temps_celsius * 9/5) + 32
print("Celsius    :", temps_celsius)
print("Fahrenheit :", np.round(temps_fahrenheit, 1))
print()

# ── Arithmetic between two arrays ────────────────────────────
humidity   = np.array([72, 65, 80, 55, 78, 85, 60])
wind_speed = np.array([12, 8, 15, 6, 11, 9, 14])

# Heat index proxy: higher temp + higher humidity = feels hotter
feels_like = temps_celsius + (humidity * 0.05) - (wind_speed * 0.1)
print("Actual temp  :", temps_celsius)
print("Feels like   :", np.round(feels_like, 1))
print()

# ── Comparison operations ─────────────────────────────────────
hot_days = temps_celsius > 30
print("Temps above 30°C:", hot_days)
print("Number of hot days:", np.sum(hot_days))   # True = 1, False = 0
```

## 🎭 Part 5 — Boolean Masking

Boolean masking lets you **filter an array by condition** —
extracting only the values that meet your criteria.
This is the NumPy equivalent of "show me only the rainy days."
**Extra reading:** Boolean masks are one of the most useful NumPy ideas. Practise them in [Topic 15C — Indexing, Slicing, and Masking](topic15c_indexing_slicing_masking.ipynb).


```python
temps = np.array([28.5, 30.1, 27.8, 31.2, 29.4, 26.5, 32.0, 25.8, 30.9])

# ── Step 1: create a boolean mask ────────────────────────────
mask_hot = temps > 30
print("Temperature array :", temps)
print("Mask (temp > 30)  :", mask_hot)
print()

# ── Step 2: apply the mask to filter values ──────────────────
hot_temps = temps[mask_hot]
print("Hot days only (>30°C) :", hot_temps)
print("Number of hot days    :", len(hot_temps))
print()

# ── Combining conditions ──────────────────────────────────────
humidity = np.array([72, 65, 80, 55, 78, 85, 60, 70, 68])

# Days that are both hot AND humid — the most uncomfortable
hot_and_humid = temps[(temps > 30) & (humidity > 70)]
print("Hot (>30°C) AND humid (>70%) :", hot_and_humid)
print()

# Days that are either very hot OR very humid
uncomfortable = temps[(temps > 31) | (humidity > 82)]
print("Very hot OR very humid       :", uncomfortable)
```

```python
# ── Masking to replace bad values ────────────────────────────
# This is a key data cleaning technique in NumPy

rainfall = np.array([12.5, -2.0, 8.1, 0.0, -0.5, 34.2, 5.5])
print("Raw rainfall (with negatives):", rainfall)

# Replace negative values with 0 (impossible rainfall)
rainfall_clean = rainfall.copy()
rainfall_clean[rainfall_clean < 0] = 0.0

print("Cleaned rainfall             :", rainfall_clean)
print(f"Fixed {np.sum(rainfall < 0)} impossible value(s)")
```

## 📊 Part 6 — Statistical Functions

NumPy gives you a complete set of statistical operations
that run on entire arrays in a single call.


```python
# Monthly average temperatures for Accra (simulated annual data)
monthly_avg = np.array([29.8, 30.2, 31.1, 30.8, 29.4, 27.6,
                        26.8, 26.5, 27.2, 28.5, 29.1, 29.5])
months = ["Jan","Feb","Mar","Apr","May","Jun",
          "Jul","Aug","Sep","Oct","Nov","Dec"]

print("ACCRA MONTHLY TEMPERATURE STATISTICS")
print("=" * 45)
print(f"  Annual mean      : {np.mean(monthly_avg):.1f} °C")
print(f"  Median           : {np.median(monthly_avg):.1f} °C")
print(f"  Std deviation    : {np.std(monthly_avg):.2f} °C")
print(f"  Hottest month    : {months[np.argmax(monthly_avg)]}  ({np.max(monthly_avg)} °C)")
print(f"  Coolest month    : {months[np.argmin(monthly_avg)]}  ({np.min(monthly_avg)} °C)")
print(f"  Range            : {np.max(monthly_avg) - np.min(monthly_avg):.1f} °C")
print(f"  Total (sum)      : {np.sum(monthly_avg):.1f} °C")
print()
print("Monthly temps:")
for m, t in zip(months, monthly_avg):
    bar = "█" * int((t - 25) * 4)
    print(f"  {m}  {t:.1f}  {bar}")
```

```python
# ── Percentiles — understanding distributions ─────────────────
temps_year = np.random.normal(29.0, 2.0, 365)   # simulate 365 days

p25  = np.percentile(temps_year, 25)
p50  = np.percentile(temps_year, 50)   # same as median
p75  = np.percentile(temps_year, 75)
p90  = np.percentile(temps_year, 90)

print("TEMPERATURE PERCENTILES (simulated year)")
print("-" * 40)
print(f"  25th percentile  : {p25:.1f} °C  (25% of days are cooler than this)")
print(f"  50th (median)    : {p50:.1f} °C")
print(f"  75th percentile  : {p75:.1f} °C  (25% of days are hotter than this)")
print(f"  90th percentile  : {p90:.1f} °C  (only 10% of days exceed this)")
```

## 🏗️ Part 7 — 2D Arrays

A 2D array has rows and columns — like a spreadsheet or a table.
In data analytics, rows typically represent observations (days, records)
and columns represent variables (temperature, rainfall, humidity).


```python
# ── Creating a 2D array ──────────────────────────────────────
# 4 days × 3 climate readings (temp, rain, humidity)
climate_table = np.array([
    [29.5, 12.0, 72],
    [31.2,  0.0, 58],
    [27.8, 34.5, 88],
    [30.1,  5.2, 65],
])

print("2D climate array:")
print(climate_table)
print()
print(f"Shape     : {climate_table.shape}  → ({climate_table.shape[0]} rows, {climate_table.shape[1]} columns)")
print(f"Total elements: {climate_table.size}")
```

```python
# ── Indexing 2D arrays ───────────────────────────────────────
print("Row 0 (first day)   :", climate_table[0])
print("Row 2 (third day)   :", climate_table[2])
print()
print("Column 0 (all temps)    :", climate_table[:, 0])
print("Column 1 (all rainfall) :", climate_table[:, 1])
print("Column 2 (all humidity) :", climate_table[:, 2])
print()
print("Row 1, Col 2 (day 2 humidity):", climate_table[1, 2])
print()
print("First 2 rows, all columns:")
print(climate_table[:2, :])
```

```python
# ── Statistical operations on 2D arrays ─────────────────────
# axis=0 → operate down columns (per variable summary)
# axis=1 → operate across rows (per observation summary)

print("Column means (axis=0) — avg per variable:")
col_means = np.mean(climate_table, axis=0)
print(f"  Avg temperature : {col_means[0]:.1f} °C")
print(f"  Avg rainfall    : {col_means[1]:.1f} mm")
print(f"  Avg humidity    : {col_means[2]:.1f} %")
print()
print("Row means (axis=1) — avg across variables per day (not meaningful here):")
print("  ", np.round(np.mean(climate_table, axis=1), 1))
```

## 🌦️ Part 8 — Analysing Accra's 2023 Climate Data

Now we apply everything — loading a real dataset and running
the full analytics pipeline using NumPy.


### Stage 1 — Data Gathering


```python
# Load the CSV using Python file handling (Topic 14 skills)
# then convert columns to NumPy arrays

FILE_PATH = "accra_climate_2023.csv"

dates         = []
temperatures  = []
rainfall      = []
humidity      = []
wind_speed    = []

with open(FILE_PATH, "r", encoding="utf-8") as f:
    header = f.readline()  # skip header

    for line in f:
        line = line.strip()
        if not line:
            continue
        parts = line.split(",")
        if len(parts) != 5:
            continue

        dates.append(parts[0])
        temperatures.append(float(parts[1]))
        rainfall.append(float(parts[2]))

        # humidity and wind may be empty string
        humidity.append(float(parts[3]) if parts[3] else None)
        wind_speed.append(float(parts[4]) if parts[4] else None)

print(f"Days loaded : {len(dates)}")
print(f"Date range  : {dates[0]}  →  {dates[-1]}")
print()
print("First 5 rows:")
for i in range(5):
    print(f"  {dates[i]}  |  {temperatures[i]}°C  |  "
          f"{rainfall[i]}mm  |  {humidity[i]}%  |  {wind_speed[i]} km/h")
```

### Stage 2 — Data Wrangling with NumPy
**Extra reading:** When missing values become `np.nan`, use [Topic 15E — Missing Data with `np.nan`](topic15e_missing_data_with_nan.ipynb) for a slower explanation.


```python
# Convert to NumPy arrays
temp_arr  = np.array(temperatures)
rain_arr  = np.array(rainfall)

# Handle None values — replace with np.nan (NumPy's missing value marker)
humid_arr = np.array([h if h is not None else np.nan for h in humidity])
wind_arr  = np.array([w if w is not None else np.nan for w in wind_speed])

print("Arrays created:")
print(f"  temp_arr  : shape {temp_arr.shape},  dtype {temp_arr.dtype}")
print(f"  rain_arr  : shape {rain_arr.shape},  dtype {rain_arr.dtype}")
print(f"  humid_arr : shape {humid_arr.shape}, dtype {humid_arr.dtype}")
print(f"  wind_arr  : shape {wind_arr.shape},  dtype {wind_arr.dtype}")
print()
print(f"Missing humidity values  : {np.sum(np.isnan(humid_arr))}")
print(f"Missing wind values      : {np.sum(np.isnan(wind_arr))}")
```

```python
# ── Fix impossible rainfall values ──────────────────────────
neg_rain = np.sum(rain_arr < 0)
print(f"Negative rainfall values found: {neg_rain}")

rain_clean = rain_arr.copy()
rain_clean[rain_clean < 0] = 0.0
print(f"After fix — minimum rainfall  : {rain_clean.min()} mm")
print()

# ── Fix outlier temperatures using IQR method ─────────────────
q1  = np.percentile(temp_arr, 25)
q3  = np.percentile(temp_arr, 75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outlier_mask = (temp_arr < lower_bound) | (temp_arr > upper_bound)
print(f"Temperature outliers (IQR method): {np.sum(outlier_mask)}")
print(f"  Bounds: {lower_bound:.1f}°C  to  {upper_bound:.1f}°C")
print(f"  Outlier values: {temp_arr[outlier_mask]}")

# Replace outliers with the median
temp_clean = temp_arr.copy()
temp_clean[outlier_mask] = np.median(temp_arr)
print(f"  Replaced with median: {np.median(temp_arr):.1f}°C")
```

### Stage 3 — EDA: Exploring the Climate Data


```python
# ── Annual summary statistics ────────────────────────────────

print("=" * 50)
print("  ACCRA CLIMATE 2023 — ANNUAL SUMMARY")
print("=" * 50)
print()
print("TEMPERATURE (°C)")
print(f"  Mean             : {np.mean(temp_clean):.1f}")
print(f"  Median           : {np.median(temp_clean):.1f}")
print(f"  Std deviation    : {np.std(temp_clean):.2f}")
print(f"  Hottest day      : {np.max(temp_clean):.1f}  (day {np.argmax(temp_clean)+1})")
print(f"  Coolest day      : {np.min(temp_clean):.1f}  (day {np.argmin(temp_clean)+1})")
print()
print("RAINFALL (mm)")
print(f"  Total annual     : {np.sum(rain_clean):.1f}")
print(f"  Average daily    : {np.mean(rain_clean):.1f}")
print(f"  Wettest day      : {np.max(rain_clean):.1f}  (day {np.argmax(rain_clean)+1})")
print(f"  Rainy days       : {np.sum(rain_clean > 0)}")
print(f"  Dry days         : {np.sum(rain_clean == 0)}")
print()
print("HUMIDITY (%)")
valid_humid = humid_arr[~np.isnan(humid_arr)]
print(f"  Mean             : {np.mean(valid_humid):.1f}")
print(f"  Range            : {np.min(valid_humid):.1f}  –  {np.max(valid_humid):.1f}")
```

```python
# ── Monthly breakdown ────────────────────────────────────────
month_names = ["Jan","Feb","Mar","Apr","May","Jun",
               "Jul","Aug","Sep","Oct","Nov","Dec"]
days_in_month = [31,28,31,30,31,30,31,31,30,31,30,31]

print(f"{'Month':<6} {'Avg Temp':>9} {'Total Rain':>11} {'Rainy Days':>11}")
print("-" * 42)

day = 0
for m_idx, (name, num_days) in enumerate(zip(month_names, days_in_month)):
    month_temp = temp_clean[day:day + num_days]
    month_rain = rain_clean[day:day + num_days]

    avg_temp   = np.mean(month_temp)
    total_rain = np.sum(month_rain)
    rainy_days = np.sum(month_rain > 0)

    rain_bar = "💧" * int(total_rain / 20)
    print(f"  {name:<4} {avg_temp:>9.1f}°C {total_rain:>10.1f}mm {rainy_days:>10} days  {rain_bar}")
    day += num_days
```

```python
# ── Seasonal analysis ────────────────────────────────────────
# Day of year ranges (approximate):
# Dry/Harmattan: days 1–59 and 300–365
# Rainy:         days 60–299

dry_mask   = (np.arange(365) < 60) | (np.arange(365) >= 300)
rainy_mask = ~dry_mask

print("SEASONAL COMPARISON")
print("-" * 45)
print(f"  Season        {'Dry':>10} {'Rainy':>12}")
print(f"  Days          {np.sum(dry_mask):>10} {np.sum(rainy_mask):>12}")
print(f"  Avg temp (°C) {np.mean(temp_clean[dry_mask]):>10.1f} {np.mean(temp_clean[rainy_mask]):>12.1f}")
print(f"  Avg rain (mm) {np.mean(rain_clean[dry_mask]):>10.1f} {np.mean(rain_clean[rainy_mask]):>12.1f}")
print(f"  Avg humidity  {np.nanmean(humid_arr[dry_mask]):>10.1f} {np.nanmean(humid_arr[rainy_mask]):>12.1f}")
```

### Stage 4 — Analysis: Answering Specific Questions


```python
# ── Q1: How many dangerously hot days (above 35°C) occurred? ──
hot_days = np.sum(temp_clean > 35)
print(f"Q1: Days above 35°C         : {hot_days}")

# ── Q2: What fraction of the year was it raining? ─────────────
rain_fraction = np.sum(rain_clean > 0) / 365 * 100
print(f"Q2: Fraction of year rainy  : {rain_fraction:.1f}%")

# ── Q3: Correlation between temperature and humidity ──────────
valid_idx = ~np.isnan(humid_arr)
corr = np.corrcoef(temp_clean[valid_idx], humid_arr[valid_idx])[0, 1]
print(f"Q3: Temp–Humidity correlation: {corr:.3f}")
direction = "negative" if corr < 0 else "positive"
print(f"    ({direction} — as temp goes up, humidity goes {'down' if corr < 0 else 'up'})")

# ── Q4: Rolling 7-day average temperature ─────────────────────
rolling_7 = np.convolve(temp_clean, np.ones(7)/7, mode='valid')
print(f"\nQ4: 7-day rolling average temp — sample (days 4-10):")
print("   ", np.round(rolling_7[:7], 1))
```

### Stage 5 — Insights


```python
# Compile all findings into a plain-language summary

dry_avg_temp   = np.mean(temp_clean[dry_mask])
rainy_avg_temp = np.mean(temp_clean[rainy_mask])
total_rainfall = np.sum(rain_clean)
rainy_day_cnt  = np.sum(rain_clean > 0)
hottest_day    = dates[np.argmax(temp_clean)]
coolest_day    = dates[np.argmin(temp_clean)]

print("=" * 65)
print("  KEY FINDINGS — Accra Climate Analysis 2023")
print("=" * 65)

findings = [
    f"Accra's 2023 mean temperature was {np.mean(temp_clean):.1f}°C, "
    f"with the hottest day on {hottest_day} and the coolest on {coolest_day}.",

    f"Total annual rainfall was {total_rainfall:.0f}mm across {rainy_day_cnt} rainy days "
    f"({rainy_day_cnt/365*100:.0f}% of the year).",

    f"The dry/harmattan season averaged {dry_avg_temp:.1f}°C — "
    f"{abs(dry_avg_temp - rainy_avg_temp):.1f}°C warmer than the rainy season "
    f"({rainy_avg_temp:.1f}°C).",

    f"Temperature and humidity show a {direction} correlation ({corr:.2f}), "
    f"consistent with Accra's tropical climate pattern.",

    f"Data quality issues found and fixed: {neg_rain} impossible rainfall "
    f"values and {np.sum(outlier_mask)} extreme temperature outliers.",
]

for i, finding in enumerate(findings, 1):
    print(f"\n  {i}. {finding}")
print("\n" + "=" * 65)
```

```python
# ⚡ LIVE CODING — Load Kumasi data
# Follow the same loading pattern as Accra above

from pathlib import Path

KUMASI_PATH = "kumasi_climate_2023.csv"
possible_paths = [
    Path(KUMASI_PATH),
    Path("data") / KUMASI_PATH,
    Path("../data") / KUMASI_PATH,
    Path("../../data") / KUMASI_PATH,
]

kumasi_file = next((path for path in possible_paths if path.exists()), None)
if kumasi_file is None:
    raise FileNotFoundError(f"Could not find {KUMASI_PATH}. Checked: {possible_paths}")

k_dates        = []
k_temperatures = []
k_rainfall     = []
k_humidity     = []
k_wind         = []

with open(kumasi_file, "r", encoding="utf-8") as f:
    header = f.readline()  # skip header

    for line in f:
        line = line.strip()
        if not line:
            continue

        parts = line.split(",")
        if len(parts) != 5:
            continue

        k_dates.append(parts[0])
        k_temperatures.append(float(parts[1]))
        k_rainfall.append(float(parts[2]))
        k_humidity.append(float(parts[3]) if parts[3] else None)
        k_wind.append(float(parts[4]) if parts[4] else None)

# Convert to arrays
k_temp_arr = np.array(k_temperatures)
k_rain_arr = np.array(k_rainfall)
k_humid_arr = np.array([h if h is not None else np.nan for h in k_humidity])
k_wind_arr = np.array([w if w is not None else np.nan for w in k_wind])

print("Kumasi data loaded.")
print(f"Source file : {kumasi_file}")
print(f"Days loaded : {len(k_dates)}")
print(f"Date range  : {k_dates[0]} -> {k_dates[-1]}")
print(f"Temperature array shape: {k_temp_arr.shape}")
print(f"Rainfall array shape   : {k_rain_arr.shape}")
```

```python
# ⚡ LIVE CODING — Clean Kumasi data
# Fix negatives and outliers the same way

# Fix impossible rainfall values
k_neg_rain = np.sum(k_rain_arr < 0)
k_rain_clean = k_rain_arr.copy()
k_rain_clean[k_rain_clean < 0] = 0.0

print(f"Negative rainfall values found: {k_neg_rain}")
print(f"After fix — minimum rainfall: {k_rain_clean.min()} mm")
print()

# Fix outlier temperatures using IQR method
k_q1 = np.percentile(k_temp_arr, 25)
k_q3 = np.percentile(k_temp_arr, 75)
k_iqr = k_q3 - k_q1
k_lower_bound = k_q1 - 1.5 * k_iqr
k_upper_bound = k_q3 + 1.5 * k_iqr

k_outlier_mask = (k_temp_arr < k_lower_bound) | (k_temp_arr > k_upper_bound)
print(f"Temperature outliers (IQR method): {np.sum(k_outlier_mask)}")
print(f"  Bounds: {k_lower_bound:.1f}°C to {k_upper_bound:.1f}°C")
print(f"  Outlier values: {k_temp_arr[k_outlier_mask]}")

k_temp_clean = k_temp_arr.copy()
k_temp_clean[k_outlier_mask] = np.median(k_temp_arr)
print(f"  Replaced with median: {np.median(k_temp_arr):.1f}°C")
```

```python
# ⚡ LIVE CODING — Kumasi summary statistics
# Print the same stats we calculated for Accra

print("=" * 50)
print("  KUMASI CLIMATE 2023 — ANNUAL SUMMARY")
print("=" * 50)
print()
print("TEMPERATURE (°C)")
print(f"  Mean             : {np.mean(k_temp_clean):.1f}")
print(f"  Median           : {np.median(k_temp_clean):.1f}")
print(f"  Std deviation    : {np.std(k_temp_clean):.2f}")
print(f"  Hottest day      : {np.max(k_temp_clean):.1f}  (day {np.argmax(k_temp_clean)+1})")
print(f"  Coolest day      : {np.min(k_temp_clean):.1f}  (day {np.argmin(k_temp_clean)+1})")
print()
print("RAINFALL (mm)")
print(f"  Total annual     : {np.sum(k_rain_clean):.1f}")
print(f"  Average daily    : {np.mean(k_rain_clean):.1f}")
print(f"  Wettest day      : {np.max(k_rain_clean):.1f}  (day {np.argmax(k_rain_clean)+1})")
print(f"  Rainy days       : {np.sum(k_rain_clean > 0)}")
print(f"  Dry days         : {np.sum(k_rain_clean == 0)}")
print()
print("HUMIDITY (%)")
k_valid_humid = k_humid_arr[~np.isnan(k_humid_arr)]
print(f"  Mean             : {np.mean(k_valid_humid):.1f}")
print(f"  Range            : {np.min(k_valid_humid):.1f} – {np.max(k_valid_humid):.1f}")
```
