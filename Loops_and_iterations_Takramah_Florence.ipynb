```python
#WAEC Grade Classifier
waec_scores = [85,72,60,50,43,91,68]
pass_count = 0

for score in waec_scores:
    if score >= 80:
        grade = "A1 | Excellent"
    elif score >= 75:
        grade = "B2 | Very Good"
    elif score >= 70:
        grade = "B3 | Good"
    elif score >= 65:
        grade = "C4 | Credit"
    elif score >= 60:
        grade = "C5 | Credit"
    elif score >= 50:
        grade = "C6 | Credit"
    else:
        grade = "F9 | Fail"

    print(f"Score: {score} | Grade: {grade}")

    if score >= 40:
        pass_count += 1

print(f"\nTotal students who passed: {pass_count}")
```

```python
#MARKET PRICE SCANNER WITH BREAK
items =["bread","chocolate", "milk","eggs"]
prices = [20.00, 25.00, 12.50, 70]

i = 0
items_checked = 0

while i < len(items):
    if prices[i] > 50:
        print(f"Warning: {items[i]} costs GH₵{prices[i]:.2f} (above limit)")
        items_checked += 1
        break
    else:
        print(f"{items[i]} - GH₵{prices[i]:.2f}")
        items_checked += 1

    i += 1

if i == len(items):
    print("All items checked")
else:
    print(f"Items checked before stopping: {items_checked}")

```

```python
#Skip the absentees
scores = [78, -1, 85, 62, -1, 90, 74]
total = 0
checked = 0
absentees = 0

for score in scores:
    if score == -1:
        print("Absent - skipped")
        absentees += 1
        continue

    total += score
    checked += 1

average = total / checked

print(f"Average score:", average)
print("Total absentees:", absentees)
```

```python
#Full payroll calculator

employee_names = ["Ama Mensah", "Kwame Boateng", "Efua Asare", "Yaw Appiah", "Akosua Darko"]
monthly_salaries = [2000, 1800, 1500, 1200, 2500]
days_worked = [22, 18, 20, 15, 22]

total_payroll = 0

for i in range(len(employee_names)):
    name = employee_names[i]
    salary = monthly_salaries[i]
    days = days_worked[i]
     # Calculate gross pay
    gross_pay = salary * (days / 22)

    # Calculate tax
    if gross_pay > 1500:
        tax = 0.05 * gross_pay
    else:
        tax = 0

    # Net pay
    net_pay = gross_pay - tax

    total_payroll += net_pay

    print(f"Name: {name}")
    print(f"Days Worked: {days}/22")
    print(f"Gross Pay: GH₵ {gross_pay:.2f}")
    print(f"Tax Deducted: GH₵ {tax:.2f}")
    print(f"Net Pay: GH₵ {net_pay:.2f}")
    print("-" * 40)

print(f"Total Payroll Cost: GH₵ {total_payroll:.2f}")
```
