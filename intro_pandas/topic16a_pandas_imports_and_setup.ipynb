## 1. The Standard Import

pandas is almost always imported with the alias `pd`:

```python
import pandas as pd
```

This means: "Load the pandas library, and let me refer to it as `pd`."


````python
import pandas as pd

print("pandas imported successfully")
print("Version:", pd.__version__)
````

You will also often see NumPy imported beside pandas:

```python
import numpy as np
import pandas as pd
```

NumPy handles fast numerical arrays. pandas builds labelled tables on top of similar ideas.


````python
import numpy as np
import pandas as pd

values = np.array([120, 250, 180])
sales = pd.Series(values, name="sales_ghs")

print(sales)
````

---
## 2. What pandas Gives Us

pandas mainly gives us two data structures:

| Structure | Shape | Think of it as |
|-----------|-------|----------------|
| `Series` | 1D | one labelled column |
| `DataFrame` | 2D | a spreadsheet/table with rows and columns |

If NumPy arrays are powerful lists of numbers, pandas DataFrames are powerful spreadsheets in Python.


---
## 3. Common Import Problems

If pandas is not installed, Python may show:

```text
ModuleNotFoundError: No module named 'pandas'
```

That means the notebook's Python environment cannot find pandas.


````python
try:
    import pandas as pd
    print("pandas is available")
except ModuleNotFoundError:
    print("pandas is not installed in this environment")
````
