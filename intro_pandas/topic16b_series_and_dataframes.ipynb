# Topic 16B — Series and DataFrames

pandas has two core data structures: `Series` and `DataFrame`.

This notebook slows down what they are, how they relate to rows and columns, and when you are working with each one.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


````python
import pandas as pd
````

---
## 1. A Series Is One Labelled Column

A Series is one-dimensional. It has values and an index.


````python
revenue = pd.Series(
    [245.50, 380.00, 120.75, 560.20],
    index=["Ama", "Kofi", "Esi", "Yaw"],
    name="daily_revenue_ghs",
)

print(revenue)
print("Name:", revenue.name)
print("Index:", list(revenue.index))
print("Values:", list(revenue.values))
````

---
## 2. A DataFrame Is a Table

A DataFrame has rows and columns. Each column is a Series.


````python
traders = pd.DataFrame({
    "name": ["Ama", "Kofi", "Esi", "Yaw"],
    "market": ["Makola", "Kejetia", "Takoradi", "Tamale"],
    "daily_revenue_ghs": [245.50, 380.00, 120.75, 560.20],
    "accepts_momo": [True, True, False, True],
})

print(traders)
print("Shape:", traders.shape)
print("Columns:", list(traders.columns))
````

---
## 3. Selecting One Column Returns a Series


````python
revenue_column = traders["daily_revenue_ghs"]

print(type(revenue_column))
print(revenue_column)
````

---
## 4. Selecting Multiple Columns Returns a DataFrame


````python
small_table = traders[["name", "daily_revenue_ghs"]]

print(type(small_table))
print(small_table)
````

---
## 5. Why Labels Matter

In NumPy, you often remember that column `0` means temperature or revenue. In pandas, the column name carries that meaning.

This makes analysis code easier to read:

```python
traders["daily_revenue_ghs"].mean()
```


````python
average_revenue = traders["daily_revenue_ghs"].mean()
print(f"Average revenue: GH₵{average_revenue:.2f}")
````
