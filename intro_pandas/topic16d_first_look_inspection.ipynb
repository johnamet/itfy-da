# Topic 16D — First Look: Inspecting a DataFrame

Before cleaning or analysis, always inspect the dataset.

This notebook focuses on the first pandas methods analysts use after loading data: `.shape`, `.head()`, `.tail()`, `.dtypes`, `.info()`, `.describe()`, and missing-value counts.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


```python
from pathlib import Path
import pandas as pd

filename = "intro_pandas/data/market_traders_500.csv"
possible_paths = [Path(filename), Path("data") / filename, Path("../data") / filename, Path("../../data") / filename]
data_path = next(path for path in possible_paths if path.exists())
df = pd.read_csv(data_path)
```

---
## 1. Shape and Columns


```python
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Column names:")
print(list(df.columns))
```

---
## 2. Preview Rows


```python
print("First rows:")
print(df.head())
print()
print("Last rows:")
print(df.tail())
```

---
## 3. Data Types

`dtypes` tells you how pandas interpreted each column.


```python
print(df.dtypes)
```

---
## 4. `info()` and Missing Values

`.info()` shows row counts, non-null counts, and data types.


```python
df.info()
print()
print("Missing values per column:")
print(df.isna().sum())
```

---
## 5. Statistical Summary


```python
print(df.describe().round(2))
```
