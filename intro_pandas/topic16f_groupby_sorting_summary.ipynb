# Topic 16F — Grouping, Sorting, and Summary Tables

`.groupby()` is one of the most important pandas tools. It lets you split data into groups, calculate summaries, and compare categories.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


```python
from pathlib import Path
import pandas as pd

filename = "intro_pandas/data/market_traders_500.csv"
possible_paths = [Path(filename), Path("data") / filename, Path("../data") / filename, Path("../../data") / filename]
data_path = next(path for path in possible_paths if path.exists())
df = pd.read_csv(data_path)
```

---
## 1. Frequency Counts


```python
print(df["product_category"].value_counts())
```

---
## 2. Sorting Rows


```python
top_revenue = df.sort_values("daily_revenue_ghs", ascending=False).head(10)
print(top_revenue[["trader_name", "market", "product_category", "daily_revenue_ghs"]])
```

---
## 3. Grouping by One Column


```python
market_avg = df.groupby("market")["daily_revenue_ghs"].mean().sort_values(ascending=False)
print(market_avg.round(2))
```

---
## 4. Multiple Aggregations

`.agg()` lets us calculate several summaries at once.


```python
category_summary = df.groupby("product_category")["daily_revenue_ghs"].agg(
    traders="count",
    avg_revenue="mean",
    total_revenue="sum",
    max_revenue="max",
).sort_values("avg_revenue", ascending=False)

print(category_summary.round(2))
```

---
## 5. Grouping by Two Columns


```python
momo_by_market = df.groupby(["market", "accepts_momo"])["trader_name"].count()
print(momo_by_market)
```
