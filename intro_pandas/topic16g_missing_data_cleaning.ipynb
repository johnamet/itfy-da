# Topic 16G — Missing Data and Cleaning in pandas

Real datasets have missing values, extra spaces, inconsistent labels, duplicates, and values that need type conversion.

This notebook gives slower practice with common pandas cleaning moves.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


```python
import pandas as pd
```

---
## 1. A Small Messy Dataset


```python
students = pd.DataFrame({
    "name": [" Ama ", "Kofi", "Esi", "Yaw", "Kofi"],
    "region": ["greater accra", "ASHANTI", "Central", "central ", "ASHANTI"],
    "score": [78, None, 91, 65, None],
    "registered": ["yes", "yes", "no", "yes", "yes"],
})

print(students)
```

---
## 2. Check Missing Values and Duplicates


```python
print("Missing values:")
print(students.isna().sum())
print()
print("Duplicate rows:", students.duplicated().sum())
```

---
## 3. Clean Text Columns


```python
students["name"] = students["name"].str.strip()
students["region"] = students["region"].str.strip().str.title()

print(students)
```

---
## 4. Fill Missing Numeric Values

Filling missing values changes the data. Always choose a method you can explain.


```python
average_score = students["score"].mean()
students["score_filled"] = students["score"].fillna(average_score)

print(students)
```

---
## 5. Convert Labels to Boolean Values


```python
students["registered_bool"] = students["registered"].map({"yes": True, "no": False})

print(students[["name", "registered", "registered_bool"]])
print(students.dtypes)
```

---
## 6. Drop Duplicates


```python
clean_students = students.drop_duplicates()

print("Before:", len(students))
print("After :", len(clean_students))
```
