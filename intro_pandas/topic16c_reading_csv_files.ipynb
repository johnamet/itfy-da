# Topic 16C — Reading CSV Files with `pd.read_csv()`

In file handling, you manually opened files, split lines, and built records. pandas can do most of that in one function: `pd.read_csv()`.

This notebook focuses on loading files safely and understanding important `read_csv` options.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


```python
from pathlib import Path
import pandas as pd
```

---
## 1. Finding the Data File

Depending on where the notebook runs, the CSV may be in the current folder or in `../data` / `../../data`.


```python
filename = "intro_pandas/data/market_traders_500.csv"
possible_paths = [
    Path(filename),
    Path("data") / filename,
    Path("../data") / filename,
    Path("../../data") / filename,
]

data_path = next((path for path in possible_paths if path.exists()), None)

if data_path is None:
    print("Could not find the file. Checked:")
    for path in possible_paths:
        print(" -", path)
else:
    print("Using:", data_path)
```

---
## 2. Load the CSV


```python
df = pd.read_csv(data_path)

print("Rows and columns:", df.shape)
print(df.head())
```

---
## 3. Useful `read_csv` Options

Common options include:

| Option | Use |
|--------|-----|
| `sep` | delimiter, such as `","` or `"	"` |
| `dtype` | force a column to load as a specific type |
| `usecols` | load only selected columns |
| `parse_dates` | convert date columns while loading |
| `na_values` | tell pandas which text values mean missing |


```python
selected = pd.read_csv(
    data_path,
    usecols=["trader_name", "market", "daily_revenue_ghs", "accepts_momo"],
    dtype={"trader_name": "string", "market": "string"},
)

print(selected.head())
print(selected.dtypes)
```
