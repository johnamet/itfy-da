# Topic 16E — Selecting and Filtering Data

Selecting means choosing columns or rows. Filtering means keeping only rows that match a condition.

These are the skills you use constantly in pandas.

**Main lesson link:** [Topic 16 — Introduction to pandas](topic16_pandas_intro.ipynb)


```python
from pathlib import Path
import pandas as pd

filename = "intro_pandas/data/market_traders_500.csv"
possible_paths = [Path(filename), Path("data") / filename, Path("../data") / filename, Path("../../data") / filename]
data_path = next(path for path in possible_paths if path.exists())
df = pd.read_csv(data_path)
```

---
## 1. Selecting Columns


```python
one_column = df["trader_name"]
multiple_columns = df[["trader_name", "market", "daily_revenue_ghs"]]

print(type(one_column))
print(one_column.head())
print()
print(type(multiple_columns))
print(multiple_columns.head())
```

---
## 2. `.loc[]` vs `.iloc[]`

- `.loc[]` selects by label/name
- `.iloc[]` selects by integer position


```python
print("loc example:")
print(df.loc[0:3, ["trader_name", "market"]])
print()
print("iloc example:")
print(df.iloc[0:4, 0:2])
```

---
## 3. Filtering with One Condition


```python
high_revenue = df[df["daily_revenue_ghs"] > 500]

print("High revenue traders:", len(high_revenue))
print(high_revenue[["trader_name", "daily_revenue_ghs"]].head())
```

---
## 4. Filtering with Multiple Conditions

Use `&` for AND, `|` for OR, and put each condition in parentheses.


```python
makola_high_momo = df[
    (df["market"] == "Makola Market") &
    (df["daily_revenue_ghs"] > 400) &
    (df["accepts_momo"] == True)
]

print("Makola high-revenue MoMo traders:", len(makola_high_momo))
print(makola_high_momo.head())
```

---
## 5. `.isin()` and `.query()`


```python
selected_regions = df[df["region"].isin(["Greater Accra", "Ashanti"])]
print("Greater Accra or Ashanti traders:", len(selected_regions))

experienced = df.query("years_trading >= 10 and daily_revenue_ghs > 400")
print("Experienced high earners:", len(experienced))
```
