## 🌉 The NumPy → Pandas Bridge

In Topic 15 you worked with NumPy arrays — fast, efficient,
but they had no column names or row labels. You had to track
which index meant which variable yourself.

Pandas solves that. A **pandas Series** is a NumPy array with a label.
A **pandas DataFrame** is a collection of Series — one per column —
all sharing the same index. Every pandas operation you will ever use
is NumPy under the hood, with labels and convenience methods on top.

Run the cell below to see this directly.
**Read more:** If `import pandas as pd` is still new, start with [Topic 16A — pandas Imports and Setup](topic16a_pandas_imports_and_setup.ipynb).


```python
import numpy as np
import pandas as pd

# The raw NumPy array — fast but no labels
revenue_array = np.array([245.50, 380.00, 120.75, 560.20, 88.40])

# Wrap it in a pandas Series — same data, now with labels
revenue_series = pd.Series(
    revenue_array,
    index=["Ama", "Kofi", "Abena", "Kwame", "Akua"],
    name="daily_revenue_ghs",
)

print("NumPy array:")
print(revenue_array)
print()
print("pandas Series (same data, with labels):")
print(revenue_series)
print()
print(f"The underlying NumPy array: {revenue_series.values}")
print(f"dtype: {revenue_series.dtype}")
print()
print("Access by label  :", revenue_series["Kofi"])
print("Access by index  :", revenue_series.iloc[1])
print("Mean             :", revenue_series.mean())
```

## 📊 Part 1 — The pandas Series

A Series is a **one-dimensional labelled array**.
Think of it as a single column from a spreadsheet —
a list of values with a name and row labels (the index).
**Extra reading:** [Topic 16B — Series and DataFrames](topic16b_series_and_dataframes.ipynb) explains these two pandas objects more slowly.


```python
import pandas as pd

# ── Creating a Series from a list ────────────────────────────
regions = pd.Series(
    ["Greater Accra", "Ashanti", "Western", "Eastern", "Central"],
    name="region",
)
print("Series from list:")
print(regions)
print(f"dtype: {regions.dtype}")
print()

# ── Creating a Series from a dictionary ──────────────────────
market_counts = pd.Series({
    "Makola Market":          120,
    "Kejetia Market":          98,
    "Kumasi Central Market":   85,
    "Cape Coast Market":       62,
    "Tamale Central Market":   43,
})
print("Series from dictionary:")
print(market_counts)
print()
print("Sorted descending:")
print(market_counts.sort_values(ascending=False))
```

```python
# ── Key Series properties ────────────────────────────────────
s = pd.Series([245.5, 380.0, 120.75, 560.2, 88.4], name="revenue")

print(f"name    : {s.name}")
print(f"dtype   : {s.dtype}")
print(f"shape   : {s.shape}")
print(f"size    : {s.size}")
print(f"index   : {list(s.index)}")
print()
print("Basic stats:")
print(f"  mean  : {s.mean():.2f}")
print(f"  max   : {s.max():.2f}")
print(f"  min   : {s.min():.2f}")
print(f"  sum   : {s.sum():.2f}")
```

## 📋 Part 2 — The pandas DataFrame

A DataFrame is a **two-dimensional labelled table** — rows and columns.
It is the central object in pandas and the most important data structure
you will use throughout this course.

Every column in a DataFrame is a Series. The DataFrame just organises
multiple Series together under shared row labels (the index).
**Extra reading:** [Topic 16B — Series and DataFrames](topic16b_series_and_dataframes.ipynb) gives extra examples of table-shaped data.


```python
# ── Creating a DataFrame from a dictionary ───────────────────
# Each key = column name, each value = list of column values

traders_dict = {
    "name":               ["Ama Mensah", "Kofi Asante", "Abena Owusu", "Kwame Boateng"],
    "market":             ["Makola Market", "Kejetia Market", "Makola Market", "Cape Coast Market"],
    "product_category":   ["Food & Produce", "Fabric & Clothing", "Electronics", "Food & Produce"],
    "daily_revenue_ghs":  [245.50, 380.00, 620.75, 88.40],
    "accepts_momo":       [True, True, False, True],
}

df = pd.DataFrame(traders_dict)
print(df)
print()
print(f"Shape  : {df.shape}   → ({df.shape[0]} rows, {df.shape[1]} columns)")
print(f"Columns: {list(df.columns)}")
```

```python
# ── Creating a DataFrame from a list of dictionaries ─────────
# This is the JSON / API format — each dict = one row

records = [
    {"name": "Ama",   "region": "Greater Accra", "revenue": 245.5},
    {"name": "Kofi",  "region": "Ashanti",        "revenue": 380.0},
    {"name": "Abena", "region": "Western",         "revenue": 620.75},
]

df2 = pd.DataFrame(records)
print("DataFrame from list of dicts:")
print(df2)
print()
print("Each column is a Series:")
print(type(df2["revenue"]))
print(df2["revenue"])
```

## 🎯 Part 5 — Selecting Columns and Rows

Knowing how to select exactly the data you need is
the foundation of all pandas work.
**Extra reading:** [Topic 16E — Selecting and Filtering Data](topic16e_selecting_and_filtering.ipynb) gives more practice with columns, `.loc[]`, `.iloc[]`, and boolean conditions.


```python
df = pd.read_csv("./data/market_traders_500.csv")

# ── Selecting a single column → returns a Series ─────────────
names = df["trader_name"]
print(type(names))
print(names.head())
print()

# ── Selecting multiple columns → returns a DataFrame ─────────
key_cols = df[["trader_name", "market", "daily_revenue_ghs"]]
print(key_cols.head())
```

```python
# ── .loc[] — label-based selection ──────────────────────────
# df.loc[row_label, column_label]

print("Single row by index label:")
print(df.loc[0])
print()

print("Row range with specific columns:")
print(df.loc[0:4, ["trader_name", "market", "daily_revenue_ghs"]])
```

```python
# ── .iloc[] — position-based selection ──────────────────────
# df.iloc[row_position, column_position]

print("First row by position:")
print(df.iloc[0])
print()

print("Rows 5–9, columns 0–2:")
print(df.iloc[5:10, 0:3])
print()

print("Last row:")
print(df.iloc[-1])
```

```python
# ── loc vs iloc — the key difference ────────────────────────
print("KEY DIFFERENCE: loc vs iloc")
print()
print("df.loc[3, 'trader_name']   → row with INDEX LABEL 3, column named 'trader_name'")
print(df.loc[3, "trader_name"])
print()
print("df.iloc[3, 0]              → row at POSITION 3, column at POSITION 0")
print(df.iloc[3, 0])
print()
print("When the index is 0,1,2,3... they give the same result.")
print("But if the index has custom labels (e.g. after filtering), they differ.")
```

## 🔍 Part 6 — Filtering Rows with Boolean Conditions

Filtering is asking pandas: *"Show me only the rows where this is true."*
It works exactly like NumPy boolean masking — but with column names.
**Extra reading:** Filtering is where many pandas mistakes happen. Practise it in [Topic 16E — Selecting and Filtering Data](topic16e_selecting_and_filtering.ipynb).


```python
df = pd.read_csv("./data/market_traders_500.csv")

# ── Single condition ─────────────────────────────────────────
makola = df[df["market"] == "Makola Market"]
print(f"Makola Market traders: {len(makola)}")
print(makola[["trader_name", "product_category", "daily_revenue_ghs"]].head())
print()

# ── Numeric condition ─────────────────────────────────────────
high_earners = df[df["daily_revenue_ghs"] > 800]
print(f"Traders earning above GH₵800/day: {len(high_earners)}")
print(high_earners[["trader_name", "market", "daily_revenue_ghs"]].head())
```

```python
# ── Multiple conditions ───────────────────────────────────────
# Use & for AND,  | for OR,  ~ for NOT
# IMPORTANT: wrap each condition in parentheses ()

# Makola traders who earn above GH₵500
makola_high = df[
    (df["market"] == "Makola Market") &
    (df["daily_revenue_ghs"] > 500)
]
print(f"Makola traders earning > GH₵500/day: {len(makola_high)}")
print(makola_high[["trader_name", "product_category", "daily_revenue_ghs"]])
print()

# Electronics OR Mobile Phones traders
tech_traders = df[
    (df["product_category"] == "Electronics") |
    (df["product_category"] == "Mobile Phones & Accessories")
]
print(f"Tech traders (Electronics + Mobile): {len(tech_traders)}")
```

```python
# ── .isin() — filter by a list of values ─────────────────────
# Much cleaner than chaining multiple == conditions

accra_ashanti = df[df["region"].isin(["Greater Accra", "Ashanti"])]
print(f"Traders in Accra and Ashanti: {len(accra_ashanti)}")
print()

# ── ~ NOT operator — exclude values ───────────────────────────
not_food = df[~(df["product_category"] == "Food & Produce")]
print(f"Non-food traders: {len(not_food)}")
print(f"Food traders:     {len(df) - len(not_food)}")
```

```python
# ── .query() — a cleaner syntax for filtering ─────────────────
# Some analysts prefer this for readability

# Traders with 10+ years experience earning above GH₵400
experienced_high = df.query("years_trading >= 10 and daily_revenue_ghs > 400")
print(f"Experienced (10+ yrs) & high earner (>GH₵400): {len(experienced_high)}")
print(experienced_high[["trader_name", "years_trading", "daily_revenue_ghs"]].head(8))
```

## 📈 Part 7 — Exploring with .value_counts() and .sort_values()

These two methods answer the most common exploratory questions:
*"How many of each?"* and *"What is the order?"*
**Extra reading:** [Topic 16F — Grouping, Sorting, and Summary Tables](topic16f_groupby_sorting_summary.ipynb) expands `.value_counts()`, `.sort_values()`, and `.groupby()`.


```python
df = pd.read_csv("./data/market_traders_500.csv")

# ── value_counts() — frequency count of a categorical column ──
print("Traders per product category:")
print(df["product_category"].value_counts())
print()

print("Traders per market:")
print(df["market"].value_counts())
print()

# Normalise to get proportions instead of counts
print("Market share (proportion):")
print(df["market"].value_counts(normalize=True).round(3))
```

```python
# ── sort_values() — sort a DataFrame by column ───────────────

# Top 10 highest-earning traders
top10 = df.sort_values("daily_revenue_ghs", ascending=False).head(10)
print("Top 10 traders by daily revenue:")
print(top10[["trader_name", "market", "product_category", "daily_revenue_ghs"]].to_string(index=False))
print()

# Sort by multiple columns: region first, then revenue descending
multi_sort = df.sort_values(
    ["region", "daily_revenue_ghs"],
    ascending=[True, False]
)
print("Sorted by region (A→Z) then revenue (high→low):")
print(multi_sort[["region", "trader_name", "daily_revenue_ghs"]].head(10))
```

## 🏪 Part 8 — Full Pipeline: Ghana Market Traders Analysis

Now we run the complete five-stage analytics pipeline
using everything we've learned in this topic.


### Stage 1 — Data Gathering


```python
df = pd.read_csv("./data/market_traders_500.csv")

print(f"Dataset loaded: {df.shape[0]} traders across {df['market'].nunique()} markets")
print(f"Date of analysis: snapshot of 2024 market activity")
print(f"Columns: {list(df.columns)}")
```

### Stage 2 — Data Wrangling
**Extra reading:** [Topic 16G — Missing Data and Cleaning](topic16g_missing_data_cleaning.ipynb) gives slower examples of cleaning text, missing values, and duplicates.


```python
# ── Check for missing values ─────────────────────────────────
print("Missing values per column:")
print(df.isnull().sum())
print()

# ── Fix trader names (strip whitespace) ──────────────────────
before = df["trader_name"].str.startswith(" ").sum()
df["trader_name"] = df["trader_name"].str.strip()
after  = df["trader_name"].str.startswith(" ").sum()
print(f"Trader names with leading whitespace: {before} → {after} after fix")
print()

# ── Standardise stall_size capitalisation ────────────────────
print("Raw stall_size values:")
print(df["stall_size"].value_counts())
df["stall_size"] = df["stall_size"].str.strip().str.title()
print()
print("Cleaned stall_size values:")
print(df["stall_size"].value_counts())
```

```python
# ── Handle missing revenue ────────────────────────────────────
missing_rev = df["daily_revenue_ghs"].isnull().sum()
print(f"Rows with missing revenue: {missing_rev}")

# Fill missing revenue with the median for that product category
df["daily_revenue_ghs"] = df.groupby("product_category")["daily_revenue_ghs"]                              .transform(lambda x: x.fillna(x.median()))

print(f"After fill: {df['daily_revenue_ghs'].isnull().sum()} missing")
print()
print("Wrangling complete ✅")
```

### Stage 3 — EDA


```python
print("MARKET TRADERS — EXPLORATORY SUMMARY")
print("=" * 55)
print()
print(df[["daily_revenue_ghs", "years_trading"]].describe().round(2))
print()

# MoMo adoption rate
momo_rate = df["accepts_momo"].mean() * 100
print(f"MoMo adoption rate : {momo_rate:.1f}%")
print()

# Revenue distribution by category
print("Average daily revenue by product category:")
cat_revenue = df.groupby("product_category")["daily_revenue_ghs"].mean().sort_values(ascending=False)
for cat, rev in cat_revenue.items():
    bar = "█" * int(rev / 40)
    print(f"  {cat:<35} GH₵{rev:>7.2f}  {bar}")
```

```python
# Revenue by market
print("Revenue summary by market:")
market_summary = df.groupby("market")["daily_revenue_ghs"].agg(
    traders="count",
    avg_revenue="mean",
    total_revenue="sum",
    max_revenue="max",
).round(2).sort_values("total_revenue", ascending=False)
print(market_summary.to_string())
```

### Stage 4 — Analysis


```python
# ── Q1: Which category has the highest average daily revenue? ─
top_category = df.groupby("product_category")["daily_revenue_ghs"].mean().idxmax()
top_revenue  = df.groupby("product_category")["daily_revenue_ghs"].mean().max()
print(f"Q1 Highest avg revenue category: {top_category} (GH₵{top_revenue:.2f}/day)")
print()

# ── Q2: Does MoMo adoption correlate with higher revenue? ─────
momo_yes = df[df["accepts_momo"] == True]["daily_revenue_ghs"].mean()
momo_no  = df[df["accepts_momo"] == False]["daily_revenue_ghs"].mean()
print(f"Q2 Avg revenue — MoMo users    : GH₵{momo_yes:.2f}")
print(f"   Avg revenue — Non-MoMo users: GH₵{momo_no:.2f}")
diff = momo_yes - momo_no
direction = "more" if diff > 0 else "less"
print(f"   MoMo users earn GH₵{abs(diff):.2f} {direction} on average")
print()

# ── Q3: Do experienced traders earn more? ─────────────────────
df["experience_band"] = pd.cut(
    df["years_trading"],
    bins=[0, 5, 10, 15, 30],
    labels=["0–5 yrs", "6–10 yrs", "11–15 yrs", "16+ yrs"]
)
exp_revenue = df.groupby("experience_band", observed=True)["daily_revenue_ghs"].mean().round(2)
print("Q3 Avg revenue by experience band:")
for band, rev in exp_revenue.items():
    print(f"   {band:<12}  GH₵{rev}")
```

### Stage 5 — Insights


```python
print("=" * 65)
print("  KEY FINDINGS — Ghana Market Traders Revenue Analysis")
print("=" * 65)

insights = [
    f"The dataset covers {len(df)} traders across "
    f"{df['market'].nunique()} major Ghanaian markets, "
    f"with an average daily revenue of GH₵{df['daily_revenue_ghs'].mean():.2f}.",

    f"Mobile Phone & Accessories traders generate the highest average "
    f"daily revenue (GH₵{df[df['product_category']=='Mobile Phones & Accessories']['daily_revenue_ghs'].mean():.2f}), "
    f"nearly double that of Food & Produce traders "
    f"(GH₵{df[df['product_category']=='Food & Produce']['daily_revenue_ghs'].mean():.2f}).",

    f"MoMo adoption stands at {momo_rate:.0f}%. Traders who accept "
    f"MoMo earn GH₵{abs(diff):.2f} {'more' if diff > 0 else 'less'} per day on average — "
    f"suggesting a {'positive' if diff > 0 else 'negative'} association between digital payments and revenue.",

    f"Kejetia and Makola markets lead in total trader revenue, "
    f"reflecting their role as Ghana's two largest urban markets.",

    f"Revenue tends to {'increase' if exp_revenue.iloc[-1] > exp_revenue.iloc[0] else 'decrease'} "
    f"with experience — traders with 16+ years earn "
    f"GH₵{exp_revenue.iloc[-1]} vs GH₵{exp_revenue.iloc[0]} for those with 0–5 years.",
]

for i, finding in enumerate(insights, 1):
    print(f"\n  {i}. {finding}")
print("\n" + "=" * 65)
```
