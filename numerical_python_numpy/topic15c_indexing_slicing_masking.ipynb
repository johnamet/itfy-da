# Topic 15C — Indexing, Slicing, and Boolean Masking

Indexing means selecting values by position. Slicing means selecting a range. Boolean masking means selecting values by condition.

These three skills appear again and again in NumPy.

**Main lesson link:** [Topic 15 — NumPy](topic15_numpy.ipynb)


```python
import numpy as np
```

---
## 1. Indexing: Selecting One Value

NumPy uses zero-based indexing, just like Python lists.


```python
temps = np.array([29.1, 30.4, 28.7, 31.2, 27.5, 29.8, 30.0])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

print("First temperature:", temps[0])
print("Fourth temperature:", temps[3])
print("Last temperature:", temps[-1])
```

---
## 2. Slicing: Selecting a Range

Slice syntax is `[start:end]`. The start is included. The end is excluded.


```python
print("First three days:", temps[0:3])
print("Weekend:", temps[5:7])
print("Every other day:", temps[::2])
```

---
## 3. Boolean Masks: Selecting by Condition

A boolean mask is an array of `True` and `False` values.


```python
hot_mask = temps > 30

print("Temperatures:", temps)
print("Hot mask    :", hot_mask)
print("Hot days    :", days[hot_mask])
print("Hot temps   :", temps[hot_mask])
```

The condition `temps > 30` is checked for every value. Only values where the mask is `True` are selected.


---
## 4. Combining Conditions

Use:

- `&` for **and**
- `|` for **or**
- `~` for **not**

Put each condition in parentheses.


```python
warm_not_extreme = (temps >= 29) & (temps <= 30.5)

print("Days:", days[warm_not_extreme])
print("Temps:", temps[warm_not_extreme])
```

---
## 5. Replacing Values with a Mask

Masks are useful for cleaning data.


```python
rainfall = np.array([12.5, -2.0, 8.1, 0.0, -0.5, 34.2])
clean_rainfall = rainfall.copy()

clean_rainfall[clean_rainfall < 0] = 0

print("Original:", rainfall)
print("Cleaned :", clean_rainfall)
```
