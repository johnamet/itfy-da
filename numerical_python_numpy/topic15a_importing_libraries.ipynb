# Topic 15A — Importing Libraries in Python

Before NumPy, most of our code used tools that Python gives us automatically: `print()`, `len()`, `sum()`, lists, dictionaries, loops, and functions.

NumPy is different. NumPy is a **library**, so we must import it before using it.

This notebook explains what libraries are, why imports exist, and how to read common import patterns such as:

```python
import numpy as np
```

**Main lesson link:** [Topic 15 — NumPy](topic15_numpy.ipynb)


---
## 1. What Is a Library?

A **library** is reusable code written by other programmers so we do not have to build everything from scratch.

Think of Python as a school bag:
- Some tools are already inside the bag, like `print()`, `len()`, and `sum()`.
- Some tools are in the school building, but you must ask for them before using them. These are standard library modules like `math`, `random`, and `statistics`.
- Some tools are from outside Python and must be installed first. NumPy is one of these.

Related words:

| Word | Meaning | Example |
|------|---------|---------|
| Module | Usually one Python file of reusable code | `math` |
| Package | A folder of modules | `numpy` |
| Library | A general name for reusable code | NumPy, pandas, Matplotlib |
| Import | The command that brings a module/package into your notebook | `import math` |


````python
# Built-in tools do not need import.
numbers = [4, 9, 16]

print(len(numbers))
print(sum(numbers))
````

---
## 2. Importing a Standard Library Module

Python includes many modules in its **standard library**. They come with Python, but they are not loaded automatically.

We import them when we need them.


````python
import math

print(math.sqrt(16))
print(math.pi)
````

Notice the pattern:

```python
import math
math.sqrt(16)
```

After `import math`, we use `math.` before the function name. This tells Python exactly where `sqrt()` came from.


---
## 3. Importing With an Alias

An **alias** is a shorter name.

NumPy is almost always imported like this:

```python
import numpy as np
```

This means: "Import the NumPy package, but let me refer to it as `np`."


````python
import numpy as np

temperatures = np.array([28.5, 30.1, 27.8])
print(temperatures)
print(type(temperatures))
````

Why use `np`?

Because NumPy code is written many times in data analysis. This is shorter:

```python
np.mean(temperatures)
```

than this:

```python
numpy.mean(temperatures)
```

The alias `np` is a community convention. When Python analysts see `np`, they expect it to mean NumPy.


---
## 4. Importing Specific Names

Sometimes we import only one function from a module.


````python
from math import sqrt

print(sqrt(25))
````

This works, but use it carefully. If many functions are imported directly, it can become hard to tell where names came from.

For data analytics, this style is usually clearer:

```python
import numpy as np
np.mean(values)
```

than this:

```python
from numpy import mean
mean(values)
```


---
## 5. Avoid `from module import *`

You may see this online:

```python
from math import *
```

It means "import everything." This is usually a bad habit in notebooks for beginners because it hides where names came from and can overwrite names you already created.

Prefer clear imports:

```python
import math
import numpy as np
```


---
## 6. What Happens When an Import Fails?

If Python cannot find a library, you may see:

```text
ModuleNotFoundError: No module named 'numpy'
```

That usually means the package is not installed in the Python environment your notebook is using.


````python
try:
    import numpy as np
    print("NumPy is available.")
    print("Version:", np.__version__)
except ModuleNotFoundError:
    print("NumPy is not installed in this environment.")
````

---
## 7. Import Habits for This Course

Use these habits unless your instructor says otherwise:

- Put imports near the top of the notebook.
- Use `import numpy as np` for NumPy.
- Keep imports clear and boring. Boring imports are good imports.
- Avoid `from module import *`.
- If an import fails, check the notebook kernel/environment.
