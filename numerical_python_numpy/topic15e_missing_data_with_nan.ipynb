# Topic 15E — Missing Data with `np.nan`

Real datasets often contain missing values. In NumPy, missing numeric values are commonly represented with `np.nan`, which means "Not a Number."

This notebook explains how to detect, ignore, and reason about missing values in numerical arrays.

**Main lesson link:** [Topic 15 — NumPy](topic15_numpy.ipynb)


```python
import numpy as np
```

---
## 1. Why Missing Values Matter

Missing is not the same as zero.

If a humidity reading is missing, writing `0` would mean "0% humidity," which is a real measurement and probably false. `np.nan` means the value is unknown.


```python
humidity = np.array([72, 58, np.nan, 80, 69])

print(humidity)
```

---
## 2. Detecting Missing Values

Use `np.isnan()` to check which values are missing.


```python
missing_mask = np.isnan(humidity)

print("Humidity    :", humidity)
print("Missing mask:", missing_mask)
print("Missing count:", np.sum(missing_mask))
```

---
## 3. Normal Statistics vs NaN-Aware Statistics

Many normal NumPy statistics return `nan` if the data contains a missing value.


```python
print("Mean with np.mean   :", np.mean(humidity))
print("Mean with np.nanmean:", np.nanmean(humidity))
```

The `nan` versions ignore missing values:

- `np.nanmean()`
- `np.nanmin()`
- `np.nanmax()`
- `np.nanmedian()`


```python
print("Minimum:", np.nanmin(humidity))
print("Maximum:", np.nanmax(humidity))
print("Median :", np.nanmedian(humidity))
```

---
## 4. Replacing Missing Values

Sometimes we need to replace missing values before modelling or exporting.

One simple method is to replace missing values with the mean of the available values.


```python
humidity_filled = humidity.copy()
mean_humidity = np.nanmean(humidity_filled)

humidity_filled[np.isnan(humidity_filled)] = mean_humidity

print("Original:", humidity)
print("Filled  :", humidity_filled)
```

Be careful: filling missing values changes the data. Always record what you did.
