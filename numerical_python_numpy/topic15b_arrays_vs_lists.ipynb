# Topic 15B — NumPy Arrays vs Python Lists

Python lists are flexible. NumPy arrays are specialised for numerical work.

This notebook helps you understand when to use each one and why NumPy is so important for data analysis.

**Main lesson link:** [Topic 15 — NumPy](topic15_numpy.ipynb)


```python
import numpy as np
```

---
## 1. Lists Can Hold Mixed Data

Python lists can hold different kinds of values at the same time.


```python
mixed_list = ["Accra", 29.5, True, 12]
print(mixed_list)
```

This flexibility is useful, but it means Python must treat each item as a general object.

NumPy arrays are designed for values that share a common type, especially numbers.


```python
temps = np.array([28.5, 30.1, 27.8, 31.2])
print(temps)
print(temps.dtype)
```

---
## 2. List Arithmetic vs Array Arithmetic

A common beginner surprise: adding a number to a list does not add that number to every item.


```python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("Array + 10:", numpy_array + 10)

try:
    print(python_list + 10)
except TypeError as error:
    print("List + 10 failed:", error)
```

To add 10 to every item in a list, we need a loop or list comprehension.

NumPy does the element-by-element operation directly.


```python
list_result = [value + 10 for value in python_list]
array_result = numpy_array + 10

print("List result :", list_result)
print("Array result:", array_result)
```

---
## 3. Shape and Data Type

NumPy arrays carry information that lists do not have in the same way:

- `shape`: the size of each dimension
- `dtype`: the type of data stored in the array
- `ndim`: the number of dimensions


```python
climate = np.array([
    [29.5, 12.0, 72],
    [31.2, 0.0, 58],
    [28.8, 5.5, 80],
])

print("Array:")
print(climate)
print("shape:", climate.shape)
print("dtype:", climate.dtype)
print("ndim :", climate.ndim)
```

---
## 4. When Should You Use Each?

| Use Python lists when... | Use NumPy arrays when... |
|--------------------------|--------------------------|
| You have mixed types | You have mostly numeric data |
| You are collecting records | You need fast calculations |
| The data is small and simple | You need statistics, masks, or 2D tables |
| You need flexible appending/removing | You need vectorised operations |

In this course, we often start with lists when reading a file, then convert the cleaned values into NumPy arrays for analysis.
