# Topic 15D — Vectorisation and Broadcasting

Vectorisation means applying an operation to many values at once. Broadcasting is NumPy's rule system for making operations work between arrays of different but compatible shapes.

These ideas are why NumPy code can be short, fast, and expressive.

**Main lesson link:** [Topic 15 — NumPy](topic15_numpy.ipynb)


```python
import numpy as np
```

---
## 1. Vectorisation: No Manual Loop Needed

With a Python list, you usually need a loop to transform every value.


```python
temps_c = np.array([28.5, 30.1, 27.8, 31.2])
temps_f = (temps_c * 9 / 5) + 32

print("Celsius   :", temps_c)
print("Fahrenheit:", temps_f)
```

NumPy applies the formula to every element in the array.


---
## 2. Broadcasting with a Single Number

When we add one number to an array, NumPy treats the number as if it should be applied to every element.


```python
rainfall_mm = np.array([0, 12.5, 34.2, 8.1])
rainfall_cm = rainfall_mm / 10

print(rainfall_cm)
```

---
## 3. Broadcasting Across Columns

Broadcasting can also work with 2D arrays.

Suppose each row is a day and each column is a climate measurement. We can subtract one set of column averages from the whole table.


```python
climate = np.array([
    [29.5, 12.0, 72],
    [31.2, 0.0, 58],
    [28.8, 5.5, 80],
    [30.1, 2.0, 69],
])

column_means = np.mean(climate, axis=0)
centered = climate - column_means

print("Column means:", column_means)
print("Centered data:")
print(centered)
```

Why did that work?

- `climate` has shape `(4, 3)`
- `column_means` has shape `(3,)`
- NumPy can match the 3 values in `column_means` to the 3 columns in `climate`

So each column gets its own value subtracted.


---
## 4. When Broadcasting Fails

Broadcasting only works when shapes are compatible.


```python
a = np.array([1, 2, 3])
b = np.array([10, 20])

try:
    print(a + b)
except ValueError as error:
    print("Broadcasting failed:", error)
```
