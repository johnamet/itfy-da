## 1. What Is a File Path?

A **file path** is the address of a file or folder.

Examples:

| Path | Meaning |
|------|---------|
| `schools_enrolment.txt` | A file in the current folder |
| `data/schools_enrolment.txt` | A file inside a `data` folder |
| `../data/schools_enrolment.txt` | Go up one folder, then enter `data` |
| `/home/johnny/Projects/ITFY-Ghana/data/schools_enrolment.txt` | An absolute path from the root of the computer |

A **relative path** depends on where your code is running from. An **absolute path** starts from the root of the computer.


````python
from pathlib import Path

current_folder = Path.cwd()
print("Current working directory:")
print(current_folder)
````

`Path.cwd()` means "current working directory." This is the folder Python uses as its starting point for relative paths.


---
## 2. Building Paths with `pathlib`

Instead of writing long path strings by hand, Python's `pathlib` library lets us build paths with `/`.


````python
project_root_guess = Path.cwd()
data_file = project_root_guess / "data" / "schools_enrolment.txt"

print(data_file)
print("Exists?", data_file.exists())
````

If `exists()` returns `False`, it does not always mean the file is gone. It may mean your path starts from the wrong folder.


---
## 3. Parent Folders

`..` means "the parent folder." With `pathlib`, the parent folder is available through `.parent`.


````python
here = Path.cwd()

print("Here       :", here)
print("Parent     :", here.parent)
print("Grandparent:", here.parent.parent)
````

In Topic 14, the path may look like this:

```python
FILE_PATH = "../../data/schools_enrolment.txt"
```

That means: start from the notebook folder, go up two folders, then enter the `data` folder.


---
## 4. Safer Path Checks

Before opening a file, you can check whether Python can see it.


````python
possible_paths = [
    Path("schools_enrolment.txt"),
    Path("data/schools_enrolment.txt"),
    Path("../../data/schools_enrolment.txt"),
]

for path in possible_paths:
    print(f"{str(path):35s} exists? {path.exists()}")
````
