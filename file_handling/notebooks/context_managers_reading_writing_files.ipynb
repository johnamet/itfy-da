# Context Managers: Reading and Writing Files

This short notebook focuses only on **context managers** for file handling.

In Python, a context manager is usually written with `with`. When we use `with open(...) as f`, Python opens the file, lets us work with it inside the indented block, and then closes it automatically when the block finishes.

Use this as extra practice for [Topic 14: File Handling & The Data Analytics Process](topic14_file_handling.ipynb).

---
## Why `with open(...)` Matters

Files are external resources. When Python opens a file, your program gets access to something outside normal Python memory. That access should be released when you are done.

`with open(...) as f` helps because it:
- closes the file automatically
- keeps file-reading or file-writing code neatly grouped
- still closes the file if an error happens inside the block

The basic pattern is:

```python
with open("file_name.txt", "r", encoding="utf-8") as f:
    content = f.read()
```

The variable `f` only represents the open file while the indented block is running.

---
## Common File Modes

| Mode | Meaning | What it does |
|------|---------|--------------|
| `"r"` | read | Opens an existing file for reading |
| `"w"` | write | Creates a new file or replaces an existing file |
| `"a"` | append | Adds new content to the end of an existing file |
| `"x"` | create | Creates a new file, but fails if the file already exists |

Use `"w"` carefully. If the file already exists, write mode overwrites it.

````python
# Setup: create a small practice file for this notebook.
# We use write mode here because we want a fresh file each time this cell runs.

PRACTICE_FILE = "context_manager_practice.txt"

with open(PRACTICE_FILE, "w", encoding="utf-8") as f:
    f.write("School,Region,Enrolment\n")
    f.write("Madina Basic,Greater Accra,421\n")
    f.write("Keta Basic,Volta,388\n")
    f.write("Ejisu Basic,Ashanti,512\n")

print(f"Created practice file: {PRACTICE_FILE}")
````

---
## Reading a Whole File

Use `.read()` when you want the entire file as one string.

This is useful for previewing a small file, but it may not be ideal for very large files because it loads everything into memory at once.

````python
with open(PRACTICE_FILE, "r", encoding="utf-8") as f:
    content = f.read()

print(content)
print("File closed after with-block:", f.closed)
````

---
## Reading Line by Line

Use `.readlines()` when the file is row-based and you want a list of lines.

Each line normally ends with a newline character, written as `\n`. Use `.strip()` to remove it before splitting or printing.

````python
with open(PRACTICE_FILE, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("Raw first line with repr():")
print(repr(lines[0]))
print()

print("Clean lines:")
for line in lines:
    print(line.strip())
````

---
## Reading and Parsing Values

A context manager handles the file safely. After reading, we still need normal Python logic to parse the data.

This practice file is comma-separated, so each row can be split with `split(",")`.

````python
records = []

with open(PRACTICE_FILE, "r", encoding="utf-8") as f:
    lines = f.readlines()

headers = lines[0].strip().split(",")

for line in lines[1:]:
    values = line.strip().split(",")
    record = dict(zip(headers, values))
    record["Enrolment"] = int(record["Enrolment"])
    records.append(record)

for record in records:
    print(record)
````

---
## Writing a File

Use write mode, `"w"`, when you want to create a new file or replace an existing one.

Inside the `with` block, call `f.write(...)`. Python does not automatically add new lines, so include `\n` where you want a line break.

````python
SUMMARY_FILE = "context_manager_summary.txt"

total_enrolment = sum(record["Enrolment"] for record in records)

with open(SUMMARY_FILE, "w", encoding="utf-8") as f:
    f.write("Practice File Summary\n")
    f.write("---------------------\n")
    f.write(f"Schools counted: {len(records)}\n")
    f.write(f"Total enrolment: {total_enrolment}\n")

print(f"Wrote summary file: {SUMMARY_FILE}")
````

---
## Appending to a File

Use append mode, `"a"`, when you want to add to the end of a file without replacing what is already there.

````python
with open(SUMMARY_FILE, "a", encoding="utf-8") as f:
    f.write("Status: reviewed in context manager notebook\n")

with open(SUMMARY_FILE, "r", encoding="utf-8") as f:
    print(f.read())
````

---
## Handling File Errors

A context manager closes a file safely after it opens. But if the file path is wrong, Python cannot open the file in the first place.

Use `try` and `except` when a missing file is possible.

````python
missing_file = "file_that_does_not_exist.txt"

try:
    with open(missing_file, "r", encoding="utf-8") as f:
        content = f.read()
except FileNotFoundError:
    print(f"Could not find: {missing_file}")
    print("Check the spelling of the file name and the folder path.")
````
