## 🗺️ Part 1 — The Data Analytics Process

Before we write a single line of code, we need a map.

Data analysis is not just "run code and get numbers." It is a thinking
process. A good analyst moves from a messy real-world question to a clear,
defensible answer that someone can use.

Every professional data analysis — whether it's a student results report,
a business revenue dashboard, or a government health survey — follows the
same five-stage pipeline. Knowing this pipeline before you start means you
always know *where you are* and *what comes next*.

For our school enrolment dataset, the journey looks like this:

| Stage | Main question | What we produce |
|-------|---------------|-----------------|
| 1. Data Gathering | Where is the data, and how do we load it? | Raw rows from `schools_enrolment.txt` |
| 2. Data Wrangling | Is the data clean enough to trust? | Standardised regions, school types, and numeric enrolment values |
| 3. EDA | What patterns or problems do we notice? | Counts, averages, ranges, and grouped summaries |
| 4. Analysis | What specific question are we answering? | Rankings, comparisons, and top schools |
| 5. Insights | What does this mean for a real person? | Plain-language findings for education decision-makers |

The stages are shown in order, but real analysis is often a loop. You might
discover a problem during EDA and go back to wrangling. You might start writing
insights and realise you need one more analysis. That back-and-forth is normal.

### Why every stage matters

Each stage protects us from a different kind of mistake.

**Skipping Data Gathering checks** means you may misunderstand the file before
you even begin. For example, if a file is tab-separated but you split on commas,
Python will not correctly separate the columns.

**Skipping Data Wrangling** is the most common beginner mistake. If your data
has inconsistent region names, missing values, or wrong data types, every
calculation you run on it will be wrong — and the error may be invisible.

**Skipping EDA** means you jump to conclusions without understanding what the
data actually looks like. You might miss outliers, unexpected patterns, or the
fact that many records are concentrated in one region.

**Skipping Analysis discipline** means you may produce interesting numbers that
do not answer the actual question. Analysis should be focused: one question,
one method, one answer.

**Skipping Insights** means you produced numbers that no one acted on. Analysis
without communication is just arithmetic.

A useful way to remember the process:

```text
Get the data -> Clean the data -> Explore the data -> Answer the question -> Explain the meaning
```

In this notebook, **no stage gets skipped.**

## 📥 Stage 1 — Data Gathering

**What it means:** Getting the raw data into your working environment.

Data can come from many places:
- A file on disk (CSV, TXT, Excel)
- A database query
- An API call over the internet
- A form or survey response
- Manual entry

Today our data source is a **text file** — `schools_enrolment.txt` —
a tab-separated record of 80 primary schools across six regions of Ghana.
This is exactly the kind of file a district education officer might
export from a government system and hand to an analyst.

### What does "separated" mean?

A plain text data file stores each row as text. To tell where one column ends
and the next column begins, the file uses a **delimiter**: a special character
that separates values.

| Format | Delimiter | Tiny example | How Python splits it |
|--------|-----------|--------------|----------------------|
| CSV: comma-separated values | comma `,` | `Ama,Central,421` | `line.split(",")` |
| TSV: tab-separated values | tab `\t` | `Ama\tCentral\t421` | `line.split("\t")` |
| Pipe-separated values | pipe `|` | `Ama|Central|421` | `line.split("|")` |
| Space-separated values | one or more spaces | `Ama Central 421` | often needs extra care because names may contain spaces |

In this notebook, the columns are separated by **tabs**. A tab is invisible
when printed normally, but Python represents it as `\t`. That is why the
first line of the file looks like this internally:

```text
School_Name	Region	District	Total_Enrolment	Male	Female	School_Type
```

A row with seven columns therefore becomes seven values after `split("\t")`.
For example:

```python
"Madina Government Basic School	Central	Mfantsiman".split("\t")
# ['Madina Government Basic School', 'Central', 'Mfantsiman']
```

Our job in Stage 1: load the file into Python, inspect its raw structure,
and convert each text row into a Python record we can work with.

Want extra practice with `with open(...)` for reading and writing files? See the companion notebook: [Context Managers: Reading and Writing Files](context_managers_reading_writing_files.ipynb).
**Extra reading:** If paths or delimiters feel unclear, read [Topic 14A — File Paths](topic14a_file_paths_and_project_structure.ipynb) and [Topic 14B — Text Files and Delimiters](topic14b_text_files_delimiters.ipynb).


````python
# ── STEP 1A: Read the raw file content ──────────────────────

# A file path is the address of the file we want Python to open.
# From this notebook folder, "../../data" means:
#   go up two folders, then enter the data folder.
FILE_PATH = "file_handling/data/schools_enrolment.txt"   # adjust path if needed

# Context management:
# "with open(...) as f" creates a temporary file object named f.
# Python opens the file at the start of the with-block and automatically
# closes it when the block ends, even if an error happens inside the block.
# This protects the file from being left open in memory.
with open(FILE_PATH, "r", encoding="utf-8") as f:
    # "r" means read mode: we are only reading the file, not changing it.
    # encoding="utf-8" tells Python how to translate bytes into text.
    raw_content = f.read()

# By the time this line runs, the with-block has ended and the file is closed.
# raw_content still holds a copy of the text we read.

# Preview the first 600 characters so we can see the structure.
# We preview instead of printing the whole file because large files can flood
# the notebook and make the important pattern harder to see.
print("RAW FILE PREVIEW (first 600 characters):")
print("-" * 50)
print(raw_content[:600])
print("-" * 50)
print(f"\nTotal characters in file: {len(raw_content):,}")
````

````python
# ── STEP 1B: Read line by line ───────────────────────────────

# .read() gave us one long string.
# .readlines() gives us a list, where each item is one line from the file.
# This is useful because data files usually mean "one row per line".
with open(FILE_PATH, "r", encoding="utf-8") as f:
    lines = f.readlines()

print(f"Total lines in file : {len(lines)}")
print(f"Header line         : {lines[0].strip()}")
print(f"\nFirst data line (raw):")

# repr() shows hidden characters such as:
#   \t  tab between columns
#   \n  newline at the end of the row
# Seeing hidden characters helps us choose the correct delimiter.
print(repr(lines[1]))
````

````python
# ── STEP 1C: Parse the header to get column names ────────────

# The first line is the header row: it contains column names, not school data.
# .strip() removes the newline at the end.
header_line = lines[0].strip()

# Because this file is tab-separated, we split on "\t".
# If this were a comma-separated CSV file, we would split on "," instead.
columns = header_line.split("\t")

print("Header line after strip():")
print(header_line)
print()
print("Column list after split('\\t'):")
print(columns)
print()
print(f"Number of columns: {len(columns)}")
print()
for i, col in enumerate(columns):
    print(f"  [{i}]  {col}")
````

````python
# ── STEP 1D: Load all rows into a list of dictionaries ───────
#    Each dictionary = one school record
#    Key = column name,  Value = cell value

# Why dictionaries?
# A list like ['Madina...', 'Central', '421'] depends on position.
# A dictionary like {'Region': 'Central'} lets us ask for fields by name,
# which makes later code easier to read and less error-prone.

data = []
skipped = 0

for line in lines[1:]:          # skip header because it contains column names
    line = line.strip()
    if not line:                # skip any blank lines
        skipped += 1
        continue

    parts = line.split("	")    # one text line -> one list of cell values

    if len(parts) != len(columns):
        # If a row has too many/few cells, it does not match the header.
        # We skip it for now rather than quietly building a broken record.
        skipped += 1
        continue

    # zip(columns, parts) pairs each column name with its matching value:
    #   ('Region', 'Central'), ('Total_Enrolment', '421'), ...
    # dict(...) turns those pairs into one dictionary record.
    record = dict(zip(columns, parts))
    data.append(record)

print(f"Records loaded : {len(data)}")
print(f"Lines skipped  : {skipped}")
print()
print("First record:")
print("-" * 40)
for key, value in data[0].items():
    print(f"  {key:20s} -> '{value}'")
````

## 🔧 Stage 2 — Data Wrangling

**What it means:** Cleaning, fixing, and reshaping the raw data
so that every value is in a consistent, usable format.

Raw data is rarely ready for analysis. People type names differently, systems
export abbreviations, and missing values often appear as blank text. Wrangling
is the stage where we make the dataset trustworthy enough for calculations.

Look at the region counts from Stage 1. You probably saw "Greater Accra",
"GREATER ACCRA", and "greater accra" counted as three different regions —
when they are the same place. Python is being literal, so we must teach it
which labels should be treated as equivalent.

We will fix **five problems** in this file:

| # | Problem | Example raw value | What to fix |
|---|---------|-------------------|-------------|
| 1 | Region names — inconsistent capitalisation & abbreviations | `GREATER ACCRA`, `Gr. Accra` | Standardise to one clean label |
| 2 | School_Type — written differently across rows | `GOVT`, `Public`, `priv` | Standardise to `Public` / `Private` |
| 3 | School names — leading/trailing whitespace | `  Keta Basic School ` | Strip extra spaces |
| 4 | Numeric columns stored as text | `"421"` | Convert to integer `421` |
| 5 | Missing or inconsistent totals | blank total, or Male + Female not matching Total | Flag for review |

A good wrangling habit: always **inspect before cleaning** and **verify after
cleaning**. You will see that pattern in each step below.
**Extra reading:** [Topic 14D — Cleaning Text, Types, and Missing Values](topic14d_cleaning_text_and_types.ipynb) gives extra examples of `.strip()`, `.lower()`, type conversion, and missing values.


````python
# ── WRANGLING STEP 1: Standardise Region names ───────────────

# First, see all unique raw values.
# set(...) removes duplicates so we can inspect the different spellings.
# sorted(...) makes the output easier to scan.
raw_regions = sorted(set(r["Region"] for r in data))
print("Unique raw Region values:")
for r in raw_regions:
    print(f"  '{r}'")
````

````python
# ── Now write the cleaning function ─────────────────────────

# A cleaning function is useful because it gives us one clear place to define
# the rule. We can test it on one value, then apply it to every record.
def standardise_region(raw):
    """Convert any region variant to its standard clean form."""
    cleaned = raw.strip().title()   # " greater accra " -> "Greater Accra"

    # Handle abbreviations and alternate spellings that .title() cannot fix.
    corrections = {
        "Gr. Accra":       "Greater Accra",
        "G. Accra":        "Greater Accra",
        "Asante":          "Ashanti",
        "Western Region":  "Western",
        "Eastern Region":  "Eastern",
        "Central Region":  "Central",
        "Volta Region":    "Volta",
    }
    return corrections.get(cleaned, cleaned)


print("Example cleaning checks:")
for example in [" greater accra ", "GREATER ACCRA", "Gr. Accra", "Eastern Region"]:
    print(f"  {example!r:18s} -> {standardise_region(example)!r}")
print()

# Apply to all records.
for record in data:
    record["Region"] = standardise_region(record["Region"])

# Verify — should now be 6 clean values.
clean_regions = sorted(set(r["Region"] for r in data))
print(f"Unique regions after cleaning: {len(clean_regions)}")
for r in clean_regions:
    print(f"  '{r}'")
````

````python
# ── WRANGLING STEP 2: Standardise School_Type ────────────────

# See the raw mess first
raw_types = sorted(set(r["School_Type"] for r in data))
print("Raw School_Type values:")
for t in raw_types:
    print(f"  '{t}'")
````

````python
def standardise_school_type(raw):
    """Map all variants to either 'Public' or 'Private'."""
    cleaned = raw.strip().lower()
    public_variants  = {"public", "govt", "government", "gov't"}
    private_variants = {"private", "pvt", "priv"}

    if cleaned in public_variants:
        return "Public"
    elif cleaned in private_variants:
        return "Private"
    else:
        return "Unknown"   # catch anything we missed


print("Example school type checks:")
for example in ["PUBLIC", "govt", " Priv ", "mission"]:
    print(f"  {example!r:10s} -> {standardise_school_type(example)!r}")
print()

for record in data:
    record["School_Type"] = standardise_school_type(record["School_Type"])

types_clean = sorted(set(r["School_Type"] for r in data))
print("Cleaned School_Type values:", types_clean)
````

````python
# ── WRANGLING STEP 3: Strip whitespace from School_Name ──────

before = sum(1 for r in data if r["School_Name"] != r["School_Name"].strip())
print(f"Names with leading/trailing whitespace: {before}")

for record in data:
    record["School_Name"] = record["School_Name"].strip()

after = sum(1 for r in data if r["School_Name"] != r["School_Name"].strip())
print(f"After cleaning: {after}")
````

````python
# ── WRANGLING STEP 4: Handle Total_Enrolment ─────────────────
#    Some values are empty strings — we can't do maths on those.

missing_count = sum(1 for r in data if r["Total_Enrolment"] == "")
print(f"Rows with missing Total_Enrolment: {missing_count}")

# Values from a text file arrive as strings, even when they look numeric.
# Example: "421" + "10" would become "42110", not 431.
# int("421") converts the text into the number 421.
print("Before conversion example:")
print(f"  value={data[0]['Total_Enrolment']!r}, type={type(data[0]['Total_Enrolment']).__name__}")

# Convert to integer — use None for missing values.
# None means "no value available". It is better than pretending a missing
# enrolment is 0, because 0 would change totals and averages.
for record in data:
    raw = record["Total_Enrolment"]
    if raw == "":
        record["Total_Enrolment"] = None
    else:
        record["Total_Enrolment"] = int(raw)

# Also convert Male and Female.
for record in data:
    record["Male"]   = int(record["Male"])
    record["Female"] = int(record["Female"])

print("\nAfter conversion example:")
print(f"  value={data[0]['Total_Enrolment']!r}, type={type(data[0]['Total_Enrolment']).__name__}")

print("\nSample after conversion:")
for r in data[:3]:
    print(f"  {r['School_Name'][:35]:35s} | Total: {r['Total_Enrolment']} | M: {r['Male']} | F: {r['Female']}")
````

````python
# ── WRANGLING STEP 5: Flag mismatched totals ─────────────────
#    Rows where Male + Female ≠ Total_Enrolment (data entry errors)

mismatches = []
for record in data:
    if record["Total_Enrolment"] is None:
        continue
    if record["Male"] + record["Female"] != record["Total_Enrolment"]:
        mismatches.append(record)

print(f"Rows with Male + Female ≠ Total_Enrolment: {len(mismatches)}")
print()
for r in mismatches:
    print(f"  {r['School_Name'][:35]:35s}")
    print(f"    Male={r['Male']}  Female={r['Female']}  "
          f"Sum={r['Male']+r['Female']}  Total_Enrolment={r['Total_Enrolment']}")
````

## 🔍 Stage 3 — Exploratory Data Analysis (EDA)

**What it means:** Getting to know your data before drawing conclusions.

EDA answers questions like:
- What does the distribution of values look like?
- Are there obvious patterns by region, type, or size?
- What is the typical value? What is the range?
- Are there any outliers or surprising values?

Think of EDA as interviewing the dataset. You are not trying to prove a point
yet; you are asking, "What is in here? What seems normal? What seems unusual?"

Examples:
- A **count** tells us how many records are in each group.
- An **average** tells us the typical size, but can be pulled upward by very large schools.
- A **range** tells us the gap between the smallest and largest values.
- A **percentage** makes groups easier to compare when totals are different.

EDA does **not** answer your final research question yet.
Its job is to make sure you understand the data well enough
to ask the right questions in Stage 4.


````python
# ── EDA 1: Basic statistics on Total_Enrolment ───────────────

# Build a clean list of numeric enrolment values.
# We exclude None because Python cannot add missing values to real numbers.
enrolments = [r["Total_Enrolment"] for r in data if r["Total_Enrolment"] is not None]

total_students = sum(enrolments)
average        = round(total_students / len(enrolments), 1)
highest        = max(enrolments)
lowest         = min(enrolments)
enrol_range    = highest - lowest

# Find schools with highest/lowest enrolment.
# next(...) returns the first record that matches the condition.
largest_school  = next(r for r in valid_records if r["Total_Enrolment"] == highest)
smallest_school = next(r for r in valid_records if r["Total_Enrolment"] == lowest)

print("ENROLMENT STATISTICS")
print("-" * 40)
print(f"  Total students       : {total_students:,}")
print(f"  Schools with data    : {len(enrolments)}")
print(f"  Average enrolment    : {average}")
print(f"  Highest enrolment    : {highest}  ({largest_school['School_Name'].strip()})")
print(f"  Lowest enrolment     : {lowest}   ({smallest_school['School_Name'].strip()})")
print(f"  Range                : {enrol_range}")
print()
print("What to notice:")
print("  The average gives one summary number, but the highest and lowest values")
print("  show how spread out school sizes are across the dataset.")
````

````python
# ── EDA 2: Enrolment by Region ───────────────────────────────

# We are building a nested dictionary:
#   region_data['Ashanti']['schools'] = number of schools in Ashanti
#   region_data['Ashanti']['total_enrolment'] = total students in Ashanti
region_data = {}

for record in valid_records:
    region = record["Region"]
    if region not in region_data:
        region_data[region] = {"schools": 0, "total_enrolment": 0}
    region_data[region]["schools"]          += 1
    region_data[region]["total_enrolment"]  += record["Total_Enrolment"]

print(f"{'Region':<20} {'Schools':>8} {'Total Enrolment':>16} {'Avg per School':>15}")
print("-" * 62)
for region, stats in sorted(region_data.items(), key=lambda x: -x[1]["total_enrolment"]):
    avg = round(stats["total_enrolment"] / stats["schools"], 1)
    print(f"  {region:<18} {stats['schools']:>8} {stats['total_enrolment']:>16,} {avg:>15}")

print("\nWhat to notice:")
print("  Total enrolment and average per school answer different questions.")
print("  A region can have a high total because it has many schools, large schools, or both.")
````

````python
# ── EDA 3: Public vs Private breakdown ───────────────────────

type_counts = {}
for record in data:
    t = record["School_Type"]
    type_counts[t] = type_counts.get(t, 0) + 1

total = sum(type_counts.values())
print("SCHOOL TYPE BREAKDOWN")
print("-" * 35)
for school_type, count in sorted(type_counts.items(), key=lambda x: -x[1]):
    pct = round((count / total) * 100, 1)
    bar = "█" * int(pct / 3)
    print(f"  {school_type:<10} {count:>4} schools  {pct:>5}%  {bar}")
````

````python
# ── EDA 4: Gender ratio across the dataset ───────────────────

total_male   = sum(r["Male"]   for r in valid_records)
total_female = sum(r["Female"] for r in valid_records)
grand_total  = total_male + total_female

male_pct   = round((total_male   / grand_total) * 100, 1)
female_pct = round((total_female / grand_total) * 100, 1)

print("GENDER DISTRIBUTION")
print("-" * 35)
print(f"  Male students   : {total_male:,}  ({male_pct}%)")
print(f"  Female students : {total_female:,}  ({female_pct}%)")
print(f"  Total students  : {grand_total:,}")
print()

# Gender parity by region
print(f"{'Region':<20} {'Male %':>8} {'Female %':>10}")
print("-" * 42)
for region in sorted(region_data.keys()):
    region_records = [r for r in valid_records if r["Region"] == region]
    m = sum(r["Male"]   for r in region_records)
    f = sum(r["Female"] for r in region_records)
    if m + f > 0:
        mp = round(m / (m + f) * 100, 1)
        fp = round(f / (m + f) * 100, 1)
        print(f"  {region:<18} {mp:>8}% {fp:>10}%")
````

## 📊 Stage 4 — Analysis

**What it means:** Answering specific questions with your clean, explored data.

EDA was about understanding. Analysis is about answering.

The difference is subtle but important:
- EDA question: "What patterns do we see by region?"
- Analysis question: "Which region has the most students enrolled?"

In analysis, we choose a specific question, write code that answers it, and then
state the answer clearly enough for someone else to use.

We will answer three analytical questions about Ghana's school enrolment data.


````python
# ── QUESTION 1: Which region has the most students enrolled? ─

print("QUESTION 1: Which region has the most students enrolled?")
print("-" * 55)

ranked = sorted(region_data.items(), key=lambda x: -x[1]["total_enrolment"])

for rank, (region, stats) in enumerate(ranked, 1):
    bar = "█" * (stats["total_enrolment"] // 800)
    print(f"  {rank}. {region:<20} {stats['total_enrolment']:>7,}  {bar}")

print()
winner = ranked[0]
print(f"  → {winner[0]} leads with {winner[1]['total_enrolment']:,} students "
      f"across {winner[1]['schools']} schools.")
````

````python
# ── QUESTION 2: Do public schools have more students than private? ──

public_enrolment  = [r["Total_Enrolment"] for r in valid_records if r["School_Type"] == "Public"]
private_enrolment = [r["Total_Enrolment"] for r in valid_records if r["School_Type"] == "Private"]

def safe_avg(lst):
    return round(sum(lst) / len(lst), 1) if lst else 0

print("QUESTION 2: Public vs Private — who has more students per school?")
print("-" * 60)
print(f"  Public  — Schools: {len(public_enrolment):>3}  "
      f"Avg enrolment: {safe_avg(public_enrolment):>7}  "
      f"Total: {sum(public_enrolment):>8,}")
print(f"  Private — Schools: {len(private_enrolment):>3}  "
      f"Avg enrolment: {safe_avg(private_enrolment):>7}  "
      f"Total: {sum(private_enrolment):>8,}")
````

````python
# ── QUESTION 3: Which schools are in the top 5 for enrolment? ──

top5 = sorted(valid_records, key=lambda x: -x["Total_Enrolment"])[:5]

print("QUESTION 3: Top 5 schools by total enrolment")
print("-" * 60)
for rank, school in enumerate(top5, 1):
    print(f"  {rank}. {school['School_Name'].strip()[:40]:<40} "
          f"{school['Region']:<15} {school['Total_Enrolment']:>6,}")
````

**What it means:** Translating your analysis into plain language
that a non-technical audience can understand and act on.

The best analysis in the world has zero impact if no one can read it.
Insights must be:

A weak insight only repeats a number: "Greater Accra has many students."
A stronger insight adds meaning: "Greater Accra has the highest enrolment,

and learning materials there."

We will also **write our clean data to a new file** — a standard
step in any real analytics workflow. Saving the cleaned file means the next
analyst does not have to repeat every cleaning step before starting their work.
**Extra reading:** [Topic 14E — File Errors and Validation](topic14e_file_errors_and_validation.ipynb) shows how to handle missing files, malformed rows, and conversion errors.


````python
# ── Write the key insights ───────────────────────────────────

insights = [
    f"The dataset covers {len(data)} primary schools across "
    f"{len(set(r['Region'] for r in data))} regions of Ghana, "
    f"with a total of {total_students:,} students enrolled.",

    f"Greater Accra and Ashanti dominate enrolment, together accounting for "
    f"the majority of all students — reflecting urban population concentration.",

    f"Public schools account for approximately "
    f"{round(len(public_enrolment)/(len(public_enrolment)+len(private_enrolment))*100)}% "
    f"of schools and serve significantly more students on average "
    f"({safe_avg(public_enrolment)} vs {safe_avg(private_enrolment)}).",

    f"Gender parity is broadly maintained across the dataset "
    f"({male_pct}% male, {female_pct}% female), though regional "
    f"variation exists and warrants further investigation.",

    f"Data quality issues were found: {len(data) - len(valid_records)} schools "
    f"had missing enrolment figures and {len(mismatches)} had "
    f"inconsistent male/female totals — these should be verified with source records.",
]

print("=" * 65)
print("  KEY FINDINGS — Ghana Primary School Enrolment Analysis")
print("=" * 65)
for i, insight in enumerate(insights, 1):
    print(f"\n  {i}. {insight}")
print("\n" + "=" * 65)
````

````python
# ── Write cleaned data to a new file ────────────────────────

OUTPUT_PATH = "schools_enrolment_clean.txt"

# We use write mode, "w", because we want to create a fresh cleaned output file.
# If the file already exists, "w" replaces it. Use this carefully in real work.
with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    # Write header. We keep this as tab-separated text, matching the source file.
    f.write("School_Name\tRegion\tDistrict\tTotal_Enrolment\tMale\tFemale\tSchool_Type\tData_Flag\n")

    for record in data:
        # Determine data quality flag.
        # This extra column helps future users know which rows need review.
        if record["Total_Enrolment"] is None:
            flag = "MISSING_ENROLMENT"
        elif record in mismatches:
            flag = "TOTAL_MISMATCH"
        else:
            flag = "OK"

        enrolment_str = "" if record["Total_Enrolment"] is None else str(record["Total_Enrolment"])

        # join(...) is the reverse of split(...): it combines a list of values
        # into one text line with tabs between the values.
        line = "\t".join([
            record["School_Name"],
            record["Region"],
            record["District"],
            enrolment_str,
            str(record["Male"]),
            str(record["Female"]),
            record["School_Type"],
            flag,
        ])
        f.write(line + "\n")

print(f"Clean file written: {OUTPUT_PATH}")
print(f"   Rows written: {len(data)}")
print(f"   Includes 'Data_Flag' column marking quality issues")
````
