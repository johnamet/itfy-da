# Topic 14C — Records, Headers, and Dictionaries

In data analysis, one row of data is often called a **record**.

This notebook explains how a header row and a data row become a Python dictionary.

**Main lesson link:** [Topic 14 — File Handling](topic14_file_handling.ipynb)


---
## 1. Header Row vs Data Row

The header row contains column names. A data row contains values.


````python
header_line = "School_Name	Region	Total_Enrolment"
data_line = "Madina Basic	Greater Accra	421"

columns = header_line.split("	")
values = data_line.split("	")

print("Columns:", columns)
print("Values :", values)
````

---
## 2. Why Dictionaries Help

Without a dictionary, you must remember positions:

```python
values[1]  # region
```

With a dictionary, you can ask by name:

```python
record["Region"]
```


````python
record = dict(zip(columns, values))

print(record)
print("Region:", record["Region"])
````

`zip(columns, values)` pairs each column name with its matching value.


````python
for pair in zip(columns, values):
    print(pair)
````

---
## 3. A List of Records

A full dataset can be represented as a list of dictionaries.


````python
lines = [
    "School_Name	Region	Total_Enrolment",
    "Madina Basic	Greater Accra	421",
    "Keta Basic	Volta	388",
    "Ejisu Basic	Ashanti	512",
]

columns = lines[0].split("	")
records = []

for line in lines[1:]:
    values = line.split("	")
    records.append(dict(zip(columns, values)))

for record in records:
    print(record)
````

---
## 4. Converting Text Values

Values read from files start as strings. If a value should be numeric, convert it.


````python
for record in records:
    record["Total_Enrolment"] = int(record["Total_Enrolment"])

total = sum(record["Total_Enrolment"] for record in records)
print("Total enrolment:", total)
````
