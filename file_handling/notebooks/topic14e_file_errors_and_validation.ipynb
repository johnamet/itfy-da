# Topic 14E — File Errors and Validation

Real file handling needs checks. Files may be missing, rows may be malformed, and numeric conversions may fail.

This notebook teaches defensive habits for reading files safely.

**Main lesson link:** [Topic 14 — File Handling](topic14_file_handling.ipynb)


---
## 1. Missing Files

If the path is wrong, Python raises `FileNotFoundError`.


````python
file_path = "missing_school_file.txt"

try:
    with open(file_path, "r", encoding="utf-8") as f:
        content = f.read()
except FileNotFoundError:
    print(f"Could not find {file_path!r}.")
    print("Check the file name and folder path.")
````

---
## 2. Validating Row Length

If a row has fewer or more values than the header, it should be reviewed.


````python
header = "School	Region	Total"
rows = [
    "Madina Basic	Greater Accra	421",
    "Broken Row	Ashanti",
    "Keta Basic	Volta	388",
]

columns = header.split("	")

for row in rows:
    parts = row.split("	")
    if len(parts) != len(columns):
        print("Skipping malformed row:", row)
        continue
    print(dict(zip(columns, parts)))
````

---
## 3. Handling Conversion Errors

`int("421")` works. `int("unknown")` fails.


````python
raw_values = ["421", "unknown", "388"]

for raw in raw_values:
    try:
        value = int(raw)
        print("Converted:", value)
    except ValueError:
        print("Could not convert:", raw)
````

---
## 4. Collecting Data Quality Issues

In real analysis, do not only skip bad rows silently. Count them or store a note.


````python
issues = []
clean_values = []

for row_number, raw in enumerate(raw_values, start=1):
    try:
        clean_values.append(int(raw))
    except ValueError:
        issues.append({"row": row_number, "value": raw, "problem": "not an integer"})

print("Clean values:", clean_values)
print("Issues:")
for issue in issues:
    print(issue)
````

---
## 5. The Defensive File Handling Pattern

A useful pattern:

```python
try:
    open the file
    read rows
    validate row length
    convert values carefully
    record issues
except FileNotFoundError:
    explain the path problem
```

The goal is not to hide errors. The goal is to make errors understandable.
