# Topic 14D — Cleaning Text, Types, and Missing Values

File handling gets data into Python. Data cleaning makes the values usable.

This notebook focuses on common cleaning moves: stripping spaces, changing case, mapping variants, converting types, and handling blanks.

**Main lesson link:** [Topic 14 — File Handling](topic14_file_handling.ipynb)


---
## 1. Extra Spaces

Human-entered data often contains spaces before or after values. Use `.strip()` to remove them.


```python
raw_name = "  Madina Basic School  "
clean_name = raw_name.strip()

print(repr(raw_name))
print(repr(clean_name))
```

---
## 2. Case Differences

Python treats `Public`, `PUBLIC`, and `public` as different strings.


```python
values = ["Public", "PUBLIC", " public ", "Govt"]

for value in values:
    print(f"{value!r:10s} -> {value.strip().lower()!r}")
```

---
## 3. Mapping Variants to Standard Labels

Cleaning often means deciding one official label for many raw variants.


```python
def standardise_school_type(raw):
    cleaned = raw.strip().lower()

    public_variants = {"public", "govt", "government", "gov't"}
    private_variants = {"private", "pvt", "priv"}

    if cleaned in public_variants:
        return "Public"
    if cleaned in private_variants:
        return "Private"
    return "Unknown"

for value in ["PUBLIC", "govt", " Priv ", "mission"]:
    print(f"{value!r:10s} -> {standardise_school_type(value)}")
```

---
## 4. Type Conversion

A number from a text file usually arrives as a string.


```python
raw_total = "421"

print(raw_total, type(raw_total))

clean_total = int(raw_total)
print(clean_total, type(clean_total))
```

---
## 5. Missing Values

A blank string `""` often means the value is missing. Do not automatically turn missing numeric values into zero.

`None` is a good beginner-friendly marker for "no value available."


```python
raw_values = ["421", "", "388"]
cleaned_values = []

for raw in raw_values:
    if raw == "":
        cleaned_values.append(None)
    else:
        cleaned_values.append(int(raw))

print(cleaned_values)
```

To calculate with the available values, skip `None`.


```python
valid_values = [value for value in cleaned_values if value is not None]

print("Valid values:", valid_values)
print("Average:", sum(valid_values) / len(valid_values))
```
