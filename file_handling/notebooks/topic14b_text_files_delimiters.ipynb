## 1. Rows and Columns in Plain Text

A table has rows and columns. A text file has characters.

To represent a table in plain text:
- each row usually gets one line
- each column is separated by a delimiter

A **delimiter** is the character that marks where one value ends and the next value begins.


```python
csv_line = "Madina Basic,Greater Accra,421"
tsv_line = "Madina Basic\tGreater Accra\t421"
pipe_line = "Madina Basic|Greater Accra|421"

print(csv_line)
print(tsv_line)
print(pipe_line)
```

---
## 2. Splitting by the Correct Delimiter

The delimiter must match the file.


```python
print("CSV split :", csv_line.split(","))
print("TSV split :", tsv_line.split("\t"))
print("Pipe split:", pipe_line.split("|"))
```

If you split on the wrong delimiter, Python cannot separate the columns correctly.


```python
print("Wrong split:")
print(tsv_line.split(","))
```

    ---
    ## 3. Hidden Characters: `	` and `
`

    Some characters affect spacing but are hard to see when printed normally.

    | Character | Meaning |
    |-----------|---------|
    | `	` | tab, often used between columns |
    | `
` | newline, used at the end of a line |

    `repr()` helps reveal hidden characters.


```python
raw_line = "School\tRegion\tTotal\n"

print("Normal print:")
print(raw_line)

print("repr print:")
print(repr(raw_line))
```

---
## 4. Joining Values Back Into a Line

`.join()` is the reverse of `.split()`.


```python
values = ["Madina Basic", "Greater Accra", "421"]

csv_output = ",".join(values)
tsv_output = "\t".join(values)

print("CSV:", csv_output)
print("TSV:", repr(tsv_output))
```
