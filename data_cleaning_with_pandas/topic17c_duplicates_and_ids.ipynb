# Topic 17C — Duplicates, IDs, and Row Uniqueness

Duplicate handling is not just deleting repeated rows. You must know which column defines a unique record.

**Main lesson link:** [Topic 17 — Data Cleaning with pandas](topic17_data_cleaning.ipynb)


```python
from pathlib import Path
import pandas as pd
import numpy as np

filename = "data_cleaning_with_pandas/data/students_messy.csv"
possible_paths = [Path(filename), Path("data") / filename, Path("../data") / filename, Path("../../data") / filename]
data_path = next(path for path in possible_paths if path.exists())
df = pd.read_csv(data_path)
print("Loaded:", data_path)
print(df.shape)
```

---
## 1. Exact Duplicate Rows


```python
print("Exact duplicate rows:", df.duplicated().sum())
```

---
## 2. Duplicate IDs

`student_id` should identify one student. Duplicate IDs may indicate repeated records, updates, or data-entry mistakes.


```python
duplicate_id_count = df["student_id"].duplicated().sum()
print("Duplicate student_id rows:", duplicate_id_count)

duplicate_records = df[df["student_id"].duplicated(keep=False)].sort_values("student_id")
print(duplicate_records[["student_id", "full_name", "region", "math_score"]].head(20))
```

---
## 3. Keep First, Last, or None?

`drop_duplicates()` has a `keep` option:

- `keep="first"`: keep the first occurrence
- `keep="last"`: keep the last occurrence
- `keep=False`: remove all duplicate groups

Choose based on the meaning of the data.


```python
keep_first = df.drop_duplicates(subset="student_id", keep="first")
keep_last = df.drop_duplicates(subset="student_id", keep="last")
remove_all_duplicates = df.drop_duplicates(subset="student_id", keep=False)

print("Original rows       :", len(df))
print("Keep first rows     :", len(keep_first))
print("Keep last rows      :", len(keep_last))
print("Remove all dup rows :", len(remove_all_duplicates))
```

---
## 4. Verify Uniqueness After Dropping


```python
df_unique = df.drop_duplicates(subset="student_id", keep="first")
print("Duplicate IDs after fix:", df_unique["student_id"].duplicated().sum())
print("student_id is unique:", df_unique["student_id"].is_unique)
```
