# Topic 17A — Data Quality Audit Before Cleaning

Cleaning should not start with random fixes. It should start with an audit: a structured check of what is wrong, where it is wrong, and how serious it is.

**Main lesson link:** [Topic 17 — Data Cleaning with pandas](topic17_data_cleaning.ipynb)


```python
from pathlib import Path
import pandas as pd
import numpy as np

filename = "data_cleaning_with_pandas/data/students_messy.csv"
possible_paths = [Path(filename), Path("data") / filename, Path("../data") / filename, Path("../../data") / filename]
data_path = next(path for path in possible_paths if path.exists())
df = pd.read_csv(data_path)
print("Loaded:", data_path)
print(df.shape)
```

---
## 1. Basic Shape and Preview

Start by asking: how many rows and columns do we have, and what does the data look like?


```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print(df.head())
```

---
## 2. Column Types and Missing Values

`.info()` gives a compact overview: column names, non-null counts, and dtypes.


```python
df.info()
print()
print("Missing values:")
print(df.isna().sum())
```

---
## 3. Duplicate Checks

Some duplicates are exact row duplicates. Others are duplicate IDs, which are usually more serious.


```python
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate student IDs:", df["student_id"].duplicated().sum())

if df["student_id"].duplicated().any():
    print(df[df["student_id"].duplicated(keep=False)].sort_values("student_id").head())
```

---
## 4. Categorical Value Checks

For text categories, inspect unique values and frequency counts. This reveals case differences, abbreviations, and spelling variants.


```python
for col in ["region", "school_type", "year_group", "gender"]:
    print("=" * 50)
    print(col)
    print(df[col].value_counts(dropna=False).head(15))
```

---
## 5. Build a Cleaning Plan

A cleaning plan turns observations into actions.


```python
cleaning_plan = {
    "full_name": "strip leading/trailing whitespace",
    "student_id": "drop duplicate IDs after reviewing records",
    "region": "standardise case and map abbreviations",
    "school_type": "map variants to Public/Private",
    "math_score": "remove percent signs, convert to numeric, set impossible scores to missing",
    "english_score/science_score": "fill missing scores with median",
    "year_group": "map variants to JHS 1/JHS 2/JHS 3",
    "gender": "map m/f variants to Male/Female",
    "date_enrolled": "parse mixed date formats with pd.to_datetime",
}

for column, action in cleaning_plan.items():
    print(f"{column:25s} -> {action}")
```
